# Chapter 11: Testing Your Code

> En este capítulo se aprende a escribir pruebas automáticas con **pytest** para comprobar que las funciones y clases hacen lo que se espera.

## Summary

Probar el código demuestra que funciona como se espera ante todos los tipos de entrada para los que fue diseñado, y permite comprobar que los cambios nuevos **no rompen** lo que ya funcionaba. Se usa la biblioteca de terceros **pytest**, que se instala con `pip`.
Una **prueba unitaria** (*unit test*) verifica un aspecto específico del comportamiento de una función; un **test case** es un conjunto de pruebas unitarias que juntas demuestran que la función se comporta bien en toda la gama de situaciones esperadas.
En pytest, las pruebas viven en archivos cuyo nombre empieza con `test_` y son funciones cuyo nombre empieza con `test_`. Cada una llama a la función que se prueba y hace una **aserción** (`assert`): una afirmación sobre una condición. Si es cierta, la prueba pasa (`.`); si no, falla (`F`). Cuando una prueba falla **no se cambia la prueba**: se arregla el código.
Las mismas ideas sirven para probar **clases**. Un **fixture** (`@pytest.fixture`) crea un recurso que se comparte entre varias funciones de prueba, para no repetir código.

|---|---|

## Installing pytest with pip

Aunque Python incluye mucha funcionalidad en su biblioteca estándar, los desarrolladores también dependen mucho de **paquetes de terceros** (*third-party packages*): bibliotecas desarrolladas fuera del lenguaje base. Algunas se terminan incorporando a la biblioteca estándar, sobre todo las que cambian poco una vez que se corrigieron sus errores iniciales.
Muchos paquetes se mantienen fuera de la biblioteca estándar para poder desarrollarlos con un calendario independiente del lenguaje, y se actualizan con más frecuencia. Es el caso de pytest y de la mayoría de las bibliotecas de la segunda mitad del libro. No hay que confiar a ciegas en cualquier paquete de terceros, pero tampoco hay que desconfiar de que mucha funcionalidad importante se implemente así.

### Updating pip

Python incluye una herramienta llamada **pip** para instalar paquetes de terceros. Como pip instala paquetes desde recursos externos, se actualiza a menudo para corregir posibles problemas de seguridad. Por eso se empieza actualizándolo. En una terminal nueva:

````python
# terminal
python -m pip install --upgrade pip
````

```
Requirement already satisfied: pip in /.../python3.11/site-packages (22.0.4)
--snip--
Successfully installed pip-22.1.2
```

- La primera parte, `python -m pip`, le dice a Python que ejecute el módulo `pip`.
- La segunda, `install --upgrade`, le dice a pip que actualice un paquete ya instalado.
- La última, `pip`, indica qué paquete se debe actualizar.

La salida muestra que la versión anterior de pip se reemplazó por la última disponible. El mismo comando sirve para actualizar cualquier paquete:

````python
# terminal
python -m pip install --upgrade package_name
````

> En Linux, pip puede no venir incluido con la instalación de Python. Si da un error al intentar actualizarlo, ver las instrucciones del Apéndice A.

### Installing pytest

Con pip al día, se instala pytest:

````python
# terminal
python -m pip install --user pytest
````

Se sigue usando el comando central `pip install`, pero ahora sin la opción `--upgrade`; en cambio, se usa **`--user`**, que le dice a Python que instale el paquete solo para el usuario actual. La salida muestra que se instaló la última versión de pytest, junto con otros paquetes de los que depende.
El mismo comando sirve para instalar muchos paquetes de terceros:

````python
# terminal
python -m pip install --user package_name
````

> Si hay problemas al ejecutar este comando, probar el mismo comando sin la opción `--user`.

### Comparison: Installing pytest with pip

| | `python -m pip install --upgrade pip` | `python -m pip install --user pytest` |

| `--upgrade` | sí | no |

| Último argumento | el paquete a actualizar (`pip`) | el paquete a instalar (`pytest`) |

| | Biblioteca estándar | Paquete de terceros |

## Testing a Function

Para aprender a probar se necesita código que probar. Esta función recibe un nombre y un apellido y devuelve el nombre completo bien formateado:

````python
# name_function.py
def get_formatted_name(first, last):
    """Generate a neatly formatted full name."""
    full_name = f"{first} {last}"
    return full_name.title()
````

La función junta el nombre y el apellido con un espacio, los capitaliza y devuelve el nombre completo. Para comprobar que funciona, un programa `names.py` que usa la función:

````python
# names.py
from name_function import get_formatted_name

print("Enter 'q' at any time to quit.")
while True:
    first = input("\nPlease give me a first name: ")
    if first == 'q':
        break
    last = input("Please give me a last name: ")
    if last == 'q':
        break

    formatted_name = get_formatted_name(first, last)
    print(f"\tNeatly formatted name: {formatted_name}.")
````

Por suerte, pytest permite **automatizar** la prueba de la salida de una función. Así se puede estar seguro de que la función funciona con los tipos de nombres para los que se escribieron pruebas.

### Unit Tests and Test Cases

Hay muchos enfoques para probar software. Uno de los más simples es la **prueba unitaria** (*unit test*): verifica que un aspecto específico del comportamiento de una función sea correcto. Un **test case** es un conjunto de pruebas unitarias que juntas demuestran que una función se comporta como debe, en toda la gama de situaciones que se espera que maneje.
Un buen test case considera todos los tipos de entrada posibles de una función e incluye pruebas para cada situación. Un test case con **cobertura completa** (*full coverage*) incluye pruebas para todas las formas posibles de usar una función. Lograrlo en un proyecto grande puede ser abrumador. A menudo basta con escribir pruebas para los comportamientos críticos del código, y buscar la cobertura completa solo si el proyecto empieza a usarse ampliamente.

### A Passing Test

Con pytest, escribir la primera prueba unitaria es bastante directo. Se escribe una función de prueba que llama a la función que se prueba y hace una afirmación sobre el valor que devuelve. Si la afirmación es correcta, la prueba **pasa**; si no, **falla**.

````python
# test_name_function.py
from name_function import get_formatted_name

def test_first_last_name():
    """Do names like 'Janis Joplin' work?"""
    formatted_name = get_formatted_name('janis', 'joplin')
    assert formatted_name == 'Janis Joplin'
````

- **El nombre del archivo importa:** debe empezar con `test_`. Cuando se le pide a pytest que ejecute las pruebas, busca cualquier archivo que empiece con `test_` y ejecuta todas las pruebas que encuentre.
- Primero se importa la función que se va a probar, `get_formatted_name()`.
- Luego se define la función de prueba, `test_first_last_name()`. Su nombre es más largo que los que se han usado, por una buena razón. Primero, las funciones de prueba deben empezar con `test` seguido de un guion bajo: cualquier función que empiece con `test_` será descubierta por pytest y ejecutada. Además, los nombres de las pruebas deben ser **largos y descriptivos**: nunca se llamará la función uno mismo, la encontrará y ejecutará pytest. El nombre debe dar una buena idea de qué comportamiento se estaba probando cuando aparezca en un reporte.
- Se llama a la función que se prueba con los argumentos `'janis'` y `'joplin'`, y se asigna el valor devuelto a `formatted_name`.
- Finalmente se hace una **aserción** (*assertion*): una afirmación sobre una condición. Aquí se afirma que el valor de `formatted_name` debe ser `'Janis Joplin'`.

### A Failing Test

¿Cómo se ve una prueba que falla? Se modifica `get_formatted_name()` para que maneje segundos nombres, pero de una forma que **rompe** la función para nombres con solo nombre y apellido. Esta versión nueva exige un segundo nombre como argumento:

````python
# name_function.py
def get_formatted_name(first, middle, last):
    """Generate a neatly formatted full name."""
    full_name = f"{first} {middle} {last}"
    return full_name.title()
````

Esta versión debería funcionar con segundos nombres, pero al probarla se ve que se rompió la función para personas con solo nombre y apellido. Al ejecutar `pytest`:

```
========================= test session starts =========================
--snip--
test_name_function.py F                                          [100%]

============================== FAILURES ===============================
________________________ test_first_last_name _________________________

    def test_first_last_name():
        """Do names like 'Janis Joplin' work?"""
>       formatted_name = get_formatted_name('janis', 'joplin')
E       TypeError: get_formatted_name() missing 1 required positional
            argument: 'last'

test_name_function.py:5: TypeError
======================= short test summary info =======================
FAILED test_name_function.py::test_first_last_name - TypeError:
    get_formatted_name() missing 1 required positional argument: 'last'
========================== 1 failed in 0.04s ==========================
```

Hay mucha información, porque cuando una prueba falla hay mucho que podría hacer falta saber:
- Lo primero es una sola **`F`**: una prueba falló.
- Luego una sección **`FAILURES`**, porque lo más importante en una ejecución son las pruebas que fallaron.
- Se ve que la función que falló fue `test_first_last_name()`.
- Un signo **`>`** indica la línea de código que causó la falla.
- La **`E`** de la línea siguiente muestra el error real: un `TypeError` por un argumento posicional obligatorio que falta, `last`.
- La información más importante se repite en un resumen corto al final, para que, al ejecutar muchas pruebas, se vea rápido cuáles fallaron y por qué.

### Responding to a Failed Test

¿Qué hacer cuando una prueba falla? Suponiendo que se están probando las condiciones correctas, una prueba que pasa significa que la función se comporta bien, y una que falla significa que hay un error en el código nuevo. Entonces, **cuando una prueba falla, no se cambia la prueba**. Si se hiciera, las pruebas podrían pasar, pero cualquier código que llame a la función como lo hace la prueba dejaría de funcionar de repente. En cambio, **se arregla el código** que causa la falla: se examinan los cambios recién hechos a la función y se averigua cómo rompieron el comportamiento deseado.
Aquí, `get_formatted_name()` antes exigía dos parámetros y ahora exige tres: el segundo nombre obligatorio rompió el comportamiento original. La mejor opción es hacer **opcional** el segundo nombre: se mueve el parámetro `middle` al final de la lista de parámetros y se le da un valor por defecto vacío, y se agrega un `if` que arma el nombre completo según haya o no segundo nombre:

````python
# name_function.py
def get_formatted_name(first, last, middle=''):
    """Generate a neatly formatted full name."""
    if middle:
        full_name = f"{first} {middle} {last}"
    else:
        full_name = f"{first} {last}"
    return full_name.title()
````

Ahora la función debería funcionar con ambos tipos de nombres. Para saber si sigue funcionando con nombres como Janis Joplin, se vuelve a ejecutar la prueba:

```
========================= test session starts =========================
--snip--
test_name_function.py .                                       [100%]

========================== 1 passed in 0.00s ==========================
```

La prueba pasa. Es ideal: la función vuelve a funcionar para nombres como Janis Joplin sin tener que probarla a mano. Arreglar la función fue más fácil porque la prueba que falló ayudó a identificar cómo el código nuevo rompió el comportamiento existente.

## Testing a Class

En la primera parte se escribieron pruebas para una sola función. Ahora se escriben para una **clase**. Como las clases se usan en muchos programas, es útil poder demostrar que funcionan correctamente. Si hay pruebas que pasan para una clase en la que se está trabajando, se puede estar seguro de que las mejoras no romperán por accidente su comportamiento actual.

### A Variety of Assertions

Hasta ahora solo se ha visto un tipo de aserción: una afirmación de que un string tiene un valor específico. Al escribir una prueba se puede afirmar **cualquier cosa que se pueda expresar como una condición**. Si la condición es `True`, como se esperaba, se confirma el supuesto sobre cómo se comporta esa parte del programa. Si es `False`, la prueba falla y se sabe que hay un problema que resolver. Las aserciones más útiles para las primeras pruebas (Table 11-1):

| Assertion | Claim |
|---|---|
| `assert a == b` | Assert that two values are equal. |
| `assert a != b` | Assert that two values are not equal. |
| `assert a` | Assert that a evaluates to True. |
| `assert not a` | Assert that a evaluates to False. |
| `assert element in list` | Assert that an element is in a list. |
| `assert element not in list` | Assert that an element is not in a list. |

Son solo algunos ejemplos: cualquier cosa que se pueda expresar como una condición puede incluirse en una prueba.

### A Class to Test

Probar una clase es parecido a probar una función, porque gran parte del trabajo consiste en probar el comportamiento de los métodos de la clase. Pero hay algunas diferencias. Una clase que ayuda a administrar encuestas anónimas, en `survey.py`:

````python
# survey.py
class AnonymousSurvey:
    """Collect anonymous answers to a survey question."""

    def __init__(self, question):
        """Store a question, and prepare to store responses."""
        self.question = question
        self.responses = []

    def show_question(self):
        """Show the survey question."""
        print(self.question)

    def store_response(self, new_response):
        """Store a single response to the survey."""
        self.responses.append(new_response)

    def show_results(self):
        """Show all the responses that have been given."""
        print("Survey results:")
        for response in self.responses:
            print(f"- {response}")
````

La clase parte con una pregunta de la encuesta y una lista vacía para guardar las respuestas. Tiene métodos para imprimir la pregunta (`show_question()`), agregar una respuesta a la lista (`store_response()`) e imprimir todas las respuestas (`show_results()`). Para crear una instancia solo hace falta una pregunta.
Un programa que usa la clase, `language_survey.py`:

````python
# language_survey.py
from survey import AnonymousSurvey

# Define a question, and make a survey.
question = "What language did you first learn to speak?"
language_survey = AnonymousSurvey(question)

# Show the question, and store responses to the question.
language_survey.show_question()
print("Enter 'q' at any time to quit.\n")
while True:
    response = input("Language: ")
    if response == 'q':
        break
    language_survey.store_response(response)

# Show the survey results.
print("\nThank you to everyone who participated in the survey!")
language_survey.show_results()
````

### Using Fixtures

En `test_survey.py` se creó una instancia nueva de `AnonymousSurvey` en cada función de prueba. En el ejemplo corto no es un problema, pero en un proyecto real con decenas o cientos de pruebas sí lo sería.
En las pruebas, un **fixture** ayuda a preparar el entorno de prueba. A menudo significa crear un recurso que usan varias pruebas. En pytest se crea un fixture escribiendo una función con el **decorador** `@pytest.fixture`. Un decorador es una directiva que se coloca justo antes de la definición de una función; Python la aplica a la función antes de ejecutarla, para alterar cómo se comporta. No hay que preocuparse si suena complicado: se pueden usar decoradores de paquetes de terceros antes de aprender a escribirlos.
Un fixture que crea una sola encuesta para usar en las dos funciones de prueba de `test_survey.py`:

````python
# test_survey.py
import pytest
from survey import AnonymousSurvey

@pytest.fixture
def language_survey():
    """A survey that will be available to all test functions."""
    question = "What language did you first learn to speak?"
    language_survey = AnonymousSurvey(question)
    return language_survey

def test_store_single_response(language_survey):
    """Test that a single response is stored properly."""
    language_survey.store_response('English')
    assert 'English' in language_survey.responses

def test_store_three_responses(language_survey):
    """Test that three individual responses are stored properly."""
    responses = ['English', 'Spanish', 'Mandarin']
    for response in responses:
        language_survey.store_response(response)

    for response in responses:
        assert response in language_survey.responses
````

- Ahora hay que importar `pytest`, porque se usa un decorador que se define en él.
- Se aplica `@pytest.fixture` a la nueva función `language_survey()`, que construye un objeto `AnonymousSurvey` y lo devuelve.
- Las definiciones de las dos funciones de prueba cambiaron: ahora cada una tiene un parámetro llamado `language_survey`. Cuando un parámetro de una función de prueba coincide con el nombre de una función con el decorador `@pytest.fixture`, el fixture se ejecuta automáticamente y su valor de retorno se pasa a la función de prueba. Aquí, `language_survey()` entrega una instancia de `language_survey` a ambas pruebas.
- No hay código nuevo en las pruebas, pero a cada una se le quitaron dos líneas: la que definía la pregunta y la que creaba el objeto `AnonymousSurvey`.

Al ejecutar el archivo, las dos pruebas siguen pasando. Serían especialmente útiles al ampliar `AnonymousSurvey` para manejar varias respuestas por persona: se podrían ejecutar para asegurarse de no haber afectado la capacidad de guardar una respuesta o una serie de respuestas individuales.

La estructura anterior parece complicada: es parte del código más abstracto que se ha visto. **No hace falta usar fixtures de inmediato**: es mejor escribir pruebas con mucho código repetido que no escribir pruebas. Cuando se hayan escrito suficientes pruebas y la repetición estorbe, existe una forma bien establecida de resolverlo. Además, en ejemplos simples como éste los fixtures no hacen el código más corto ni más simple, pero en proyectos con muchas pruebas, o cuando hacen falta muchas líneas para armar un recurso usado en varias pruebas, pueden mejorar drásticamente el código de prueba.
Para escribir un fixture: se escribe una función que genera el recurso usado por varias pruebas, se le agrega el decorador `@pytest.fixture` y se agrega el nombre de esa función como parámetro de cada función de prueba que usa el recurso. Las pruebas quedan más cortas y más fáciles de escribir y mantener.

### Comparison: Testing a Class

| | Probar una función | Probar una clase |
|---|---|---|
| Qué se llama en la prueba | la función con ciertos argumentos | un método, **después de crear una instancia** |
| Qué se afirma | el valor devuelto | el estado de la instancia (por ejemplo `'English' in language_survey.responses`) |
| Ejemplo | `get_formatted_name('janis', 'joplin')` | `language_survey.store_response('English')` |

| Aserción | Cuándo usarla | Ejemplo |
|---|---|---|
| `assert a == b` | un valor debe ser exactamente otro | `assert formatted_name == 'Janis Joplin'` |
| `assert element in list` | un elemento debe estar guardado | `assert 'English' in language_survey.responses` |
| `assert a` | un valor debe evaluarse como `True` | `assert language_survey.responses` (ya se guardó algo: la lista no está vacía) |
| `assert not a` | algo debe ser `False` o estar vacío | `assert not language_survey.responses` (antes de guardar nada) |

| | Sin fixture | Con fixture |
|---|---|---|
| Quién crea la instancia | cada función de prueba | una función con `@pytest.fixture` |
| Líneas repetidas | la pregunta y `AnonymousSurvey(question)` en cada prueba | ninguna |
| Parámetro de la función de prueba | ninguno | el nombre del fixture (`language_survey`) |
| Necesita `import pytest` | no | sí |
| Cuándo usarlo | pocas pruebas, o al empezar | cuando la repetición estorba (muchas pruebas o recursos largos de armar) |

| | `pytest` | `pytest test_survey.py` |
|---|---|---|
| Qué ejecuta | todas las pruebas de la carpeta actual | solo las pruebas de ese archivo |
