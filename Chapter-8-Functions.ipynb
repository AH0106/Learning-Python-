## Defining a Function

Aquí se muestra una función llamada `greet_user()` que imprime un saludo

```python
def greet_user(): 
    """Display a simple greeting"""
    print("Hello!")

greet_user()
```

### Passing Information to a Function

Si se modifica la función `greet_user()` puede saludar a un usuario por su nombre.

Para que la función haga esto, se ingresa `username` dentro del parentesis que define a la función.

Añadiendo `username` ahí, se permite a la función aceptar cualquier valor de `username` que se especifique. 

```python
def greet_user(username): 
    """Display a simple greeting"""
    print(f"Hello, {username.title()}")

greet_user("Andrés")
```

### Arguments and Parameters

La variable `username` en la definición de `greet_user` es un ejemplo de `parametro`: Una pieza de información que necesita la función para hacer su trabajo. 

Un `argumento` es una pieza de información que es pasada a la llamada de una función.

## Passing Arguments

### Positional Argument

Cuando se llama a una función, Python debe hacer coincidir cada argumento en la llamada de la función con el parametro en la definición de la función.

Considerar una función que muestra información de mascotas. 

```python
def describe_pet(animal_type, pet_name) :
    """Display information about a pet"""
    print(f"\nI have a {animal_type}")
    print(f"\nMy {animal_type}'s name is {pet_name.title()}")

describe_pet('dog', 'martin')
```

### Multiple Function Calls

Se puede llamar una función tantas veces como se necesite. Describiendo un segundo animal, diferentes mascotas requieren solo una llamada más de la función `describe_pet`

```python
def describe_pet(animal_type, pet_name) :
    """Display information about a pet"""
    print(f"\nI have a {animal_type}")
    print(f"My {animal_type}'s name is {pet_name.title()}")

describe_pet('dog', 'martin')
describe_pet('cat', 'gary')
```

## Default Values

Cuando se escribe una función, se puede definir un `default value` por cada parametro. 

Si un argumento para un parametro se pasa al llamar una función, Python usa el valor del argumento.

Si no, usa el valor default del parametro.

```python
def describe_pet(pet_name, animal_type = 'dog') :
    """Display information about a pet"""
    print(f"\nI have a {animal_type}")
    print(f"My {animal_type}'s name is {pet_name.title()}")

describe_pet('Martin')

describe_pet('gary', 'cat')
```

## Return Values

Una función no siempre tiene que desplegar un output directamente. En cambio, puede procesar algun dato y luego retorna un valor o un conjunto de valores. 

El valor que retorna la función se llama `return value`. 

### Returning a Simple Value

Una función que toma el primer nombre y el apellido, y retorna el nombre completo

```python
def get_formatted_name(first_name, last_name): 
    """Return a full name, neatly formatted."""
    full_name = f"{first_name} {last_name}"
    return full_name.title()

musician = get_formatted_name('Gary', 'Moore')
print(musician)
```

## Making a Argument Optional

Algunas veces tiene sentido hacer un argumento opcional, para que la persona que use la función pueda elegir si provee información extra solo si esta quiere.

Por ejemplo, si modificamos la función `get_formatted_name()` para que pueda tener un segundo nombre:

```python
def get_formatted_name(first_name, middle_name ,last_name): 
    """Return a full name, neatly formatted."""
    full_name = f"{first_name} {middle_name} {last_name}"
    return full_name.title()

musician = get_formatted_name('Andrés', 'Eduardo', 'Hormazabal')
print(musician)
```

Pero los segundos nombres no siempre se necesitan, y esta función asi como esta escrita, no funcionara si se intenta llamar solo con el primer nombre y el apellido. 

Para hacer el segundo nombre opcional, se le da al argumento `middle_name` un valor por default vacio, y se ignora el argumento al menos que el usuario provea el valor.

```python
def get_formatted_name(first_name, last_name, middle_name = ''): 
    """Return a full name, neatly formatted."""
    if middle_name:
        full_name = f"{first_name} {middle_name} {last_name}"
    else:
        full_name = f"{first_name} {last_name}"
    
    return full_name.title()

musician = get_formatted_name('jimi', 'hendrix')
print(musician)

musician = get_formatted_name('john', 'hooker', 'lee')
print(musician)
```

## Returning a Dictionary

Las funciones pueden retornar cualquier tipo de valor que se necesite, incluyendo listas y diccionarios.

```python
def build_person(first_name, last_name):
    """Return a dictionary of information about a person"""

    person = {
        'first': first_name,
        'last' : last_name
    }
    return person

musician = build_person('gary', 'moore')
print(musician)
```

> Esta función se puede expandir para aceptar valores opcionales como segundo nombre, edad, ocupación, o cualquier otra información que se quiera guardar de una persona. 

```python
def build_person(first_name, last_name, age = None):
    """Return a dictionary of information about a person"""

    person = {
        'first': first_name,
        'last' : last_name,
    }
    if age:
        person['age'] = age

    return person

musician = build_person('gary', 'moore', 27)
print(musician)
```

## Passing a List

Cuando se pasa una lista a una función, **la función tiene acceso directo a los contenidos de la lista**.

Digamos que tenemos una lista de usuarios y se quiere imprimir un saludo a cada uno de estos

```python
def greet_users(names): 
    """Print a simple greeting to each user in the list."""
    for name in names:
        msg = f"Hello, {name.title()}"
        print(msg)

usernames = ['hanna', 'ty', 'margot']
greet_users(usernames)
```

## Modifying a List in a function

Cuanso se le pasa una lista a una función, esta puede modificar la lista.

Considere una empresa que crea modelos impresos en 3D, a partir de diseños enviados por los ususarios. Los diseños pendientes de impresión se almacenan en una lista, y una vez impresos, se trasladan a una lista independiente.

```python
# Start with some designs that need to be printed
unprinted_designs = ['phone case', 'robot pendant', 'dodecahedron']
completed_models = []

def print_models(unprinted_designs, completed_models): 
    """Simulate printing each design, until none are left.
    Move each desing to completed_models after printing"""
    while unprinted_designs:
        current_desing = unprinted_designs.pop()
        print(f"Printing model: {current_desing}")
        completed_models.append(current_desing)

def show_completed_models(completed_models):
    """Display all completed models."""
    print("\nThe following models have been printed: ")
    for completed_model in completed_models:
        print(completed_model)


print_models(unprinted_designs, completed_models)
show_completed_models(completed_models)
```
