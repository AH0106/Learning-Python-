## Summary

La función `input()` pausa el programa, muestra un mensaje (*prompt*) y espera que el usuario escriba; **todo lo que se ingresa llega como string**, así que para usarlo como número hay que convertirlo con `int()`. El operador **módulo** (`%`) da el resto de una división y sirve, por ejemplo, para saber si un número es par o impar.
Un `while` repite un bloque **mientras una condición sea verdadera**, a diferencia del `for`, que recorre una colección. Se puede controlar con una condición directa, con un **flag** (una variable `True`/`False` que indica si el programa sigue activo), con `break` (salir del loop de inmediato) y con `continue` (volver al inicio del loop).
Todo `while` necesita una forma de terminar; si no, se produce un **loop infinito** (se detiene con `Ctrl-C`).
Combinado con listas y diccionarios, el `while` permite mover elementos de una lista a otra, eliminar todas las apariciones de un valor y llenar un diccionario con las respuestas del usuario.

## How the input() Function Works

La función `input()` pausa el programa y espera que el usuario escriba algo. Cuando Python recibe la respuesta, la asigna a una variable para trabajar con ella.
Por ejemplo, este programa pide un texto y luego lo muestra de vuelta:

> Estas celdas con `input()` hay que ejecutarlas y escribir la respuesta en el cuadro que aparece (en VS Code aparece arriba). Por eso no tienen salida guardada.

````python
message = input("Tell me something, and I will repeat it back to you: ")
print(message)
````

`input()` recibe un argumento: el **prompt**, el mensaje que se le muestra al usuario para que sepa qué tipo de información ingresar. El programa espera la respuesta y continúa cuando el usuario presiona ENTER.

### Writing Clear Prompts

Cada vez que se use `input()` se debe incluir un prompt claro, que diga exactamente qué información se espera. Conviene dejar **un espacio al final** del prompt (después de los dos puntos) para separar el mensaje de la respuesta del usuario.

````python
name = input("Please enter your name: ")
print(f"\nHello, {name}!")
````

Cuando el prompt es más largo que una línea, se puede asignar a una variable y pasar esa variable a `input()`. Así se arma el mensaje en varias líneas y la instrucción `input()` queda limpia:

````python
prompt = "If you share your name, we can personalize the messages you see."
prompt += "\nWhat is your first name? "

name = input(prompt)
print(f"\nHello, {name}!")
````

El operador `+=` toma el string que ya estaba en `prompt` y le agrega el nuevo al final. Es una forma de construir un string de varias líneas.

### Using int() to Accept Numerical Input

Cuando se usa `input()`, Python interpreta **todo** lo que escribe el usuario como un string. Por ejemplo, si el usuario escribe `21`, la variable vale `'21'` (con comillas). Si solo se va a imprimir, funciona bien, pero si se intenta usar como número, hay un error:

```python
age = input("How old are you? ")
# How old are you? 21
age >= 18
# TypeError: '>=' not supported between instances of 'str' and 'int'
```

Python no puede comparar el string `'21'` con el número `18`. Se soluciona con la función `int()`, que convierte el texto en un número:

````python
age = "21"        # lo que devuelve input() al escribir 21
age = int(age)
print(age >= 18)
````

Ejemplo: un programa que revisa si una persona tiene la altura necesaria para subir a una montaña rusa.

````python
height = input("How tall are you, in inches? ")
height = int(height)

if height >= 48:
    print("\nYou're tall enough to ride!")
else:
    print("\nYou'll be able to ride when you're a little older.")
````

El programa puede comparar `height` con `48` porque antes se convirtió el valor a número. Cuando se usa lo ingresado en cálculos o comparaciones numéricas, **siempre** hay que convertirlo primero.

### The Modulo Operator

El operador **módulo** (`%`) divide un número por otro y devuelve el **resto**:

````python
print(4 % 3)
print(5 % 3)
print(6 % 3)
print(7 % 3)
````

No dice cuántas veces cabe un número en otro, solo cuál es el resto. Cuando un número es divisible por otro, el resto es `0`, y eso permite saber si un número es par o impar:

````python
number = input("Enter a number, and I'll tell you if it's even or odd: ")
number = int(number)

if number % 2 == 0:
    print(f"\nThe number {number} is even.")
else:
    print(f"\nThe number {number} is odd.")
````

Los números pares siempre son divisibles por dos, así que si el módulo de un número con `2` es `0` (`number % 2 == 0`), el número es par. Si no, es impar.

## Introducing while Loops

El `for` toma una colección de elementos y ejecuta un bloque una vez por cada elemento. En cambio, el `while` se ejecuta **mientras** una condición sea verdadera.

### The while Loop in Action

Un `while` puede contar una serie de números. Este cuenta de 1 a 5:

````python
current_number = 1
while current_number <= 5:
    print(current_number)
    current_number += 1
````

- Se parte con `current_number = 1`.
- El `while` sigue corriendo mientras `current_number` sea menor o igual a `5`.
- Dentro del loop se imprime el valor y se le suma `1` con `current_number += 1` (que es lo mismo que `current_number = current_number + 1`).
- Cuando `current_number` pasa de `5`, el loop se detiene.

Los programas de uso diario tienen `while` en alguna parte. Por ejemplo, un juego necesita un `while` para seguir corriendo mientras el jugador quiera jugar y para parar cuando éste lo pida.

### Letting the User Choose When to Quit

Se puede hacer que `parrot.py` corra mientras el usuario quiera, poniendo casi todo el programa dentro de un `while`. Se define un **valor para salir** y el programa sigue mientras el usuario no lo ingrese:

````python
prompt = "\nTell me something, and I will repeat it back to you:"
prompt += "\nEnter 'quit' to end the program. "

message = ""
while message != 'quit':
    message = input(prompt)
    print(message)
````

Se define un prompt que le dice al usuario sus dos opciones: ingresar un mensaje o ingresar el valor para salir (`'quit'`). Luego se crea la variable `message` para guardar lo que ingrese.
`message` se define como un string vacío (`""`) para que Python tenga algo que comparar la primera vez que llega a la línea del `while`. Sin un valor inicial no podría hacer la comparación y el programa no seguiría.

Este programa funciona, pero imprime la palabra `'quit'` como si fuera un mensaje. Se arregla con un `if`:

````python
prompt = "\nTell me something, and I will repeat it back to you:"
prompt += "\nEnter 'quit' to end the program. "

message = ""
while message != 'quit':
    message = input(prompt)

    if message != 'quit':
        print(message)
````

### Using a Flag

En el ejemplo anterior el programa hacía algo mientras una condición era verdadera. Pero en programas más complejos hay **muchos eventos** que pueden detener el programa.
Por ejemplo, en un juego, la partida puede terminar cuando el jugador se queda sin naves, se acaba el tiempo o se destruyen las ciudades que debía proteger. Si hay muchas condiciones, ponerlas todas en un solo `while` se vuelve complicado.

Para estos casos se define una variable que indica si el programa completo está activo. Esta variable se llama **flag** (bandera) y actúa como una señal para el programa: se escribe el programa para que corra mientras el flag sea `True` y se detenga cuando cualquier evento lo ponga en `False`. Así el `while` solo revisa una condición, y las demás pruebas quedan ordenadas en el resto del programa.

````python
prompt = "\nTell me something, and I will repeat it back to you:"
prompt += "\nEnter 'quit' to end the program. "

active = True
while active:
    message = input(prompt)

    if message == 'quit':
        active = False
    else:
        print(message)
````

Se pone `active` en `True` para que el programa parta activo. Así el `while` no necesita hacer ninguna comparación; la lógica está en otras partes. Mientras `active` siga en `True`, el loop continúa.
Dentro del `while`, si el usuario ingresa `'quit'`, `active` pasa a `False` y el loop termina; si ingresa cualquier otra cosa, se imprime.
La salida es la misma que en el ejemplo anterior, pero ahora es fácil agregar más pruebas (por ejemplo con `elif`) para otros eventos que pongan `active` en `False`.

### Using break to Exit a Loop

Para salir de un `while` de inmediato, sin ejecutar el resto del código del loop y sin importar el resultado de ninguna condición, se usa `break`. Es una instrucción que controla el flujo del programa: decide qué líneas se ejecutan y cuáles no.
Por ejemplo, un programa que pregunta por ciudades visitadas y se detiene apenas el usuario ingresa `'quit'`:

````python
prompt = "\nPlease enter the name of a city you have visited:"
prompt += "\n(Enter 'quit' when you are finished.) "

while True:
    city = input(prompt)

    if city == 'quit':
        break
    else:
        print(f"I'd love to go to {city.title()}!")
````

Un loop que parte con `while True` corre para siempre, a menos que llegue a un `break`. Cuando el usuario ingresa `'quit'`, se ejecuta el `break` y Python sale del loop.

> Se puede usar `break` en cualquier tipo de loop. Por ejemplo, para salir de un `for` que recorre una lista o un diccionario.

### Using continue in a Loop

En vez de salir del loop sin ejecutar el resto del código, `continue` **vuelve al inicio del loop** según el resultado de una condición. Por ejemplo, un loop que cuenta del 1 al 10 pero imprime solo los impares:

````python
current_number = 0
while current_number < 10:
    current_number += 1
    if current_number % 2 == 0:
        continue

    print(current_number)
````

Se parte con `current_number = 0`. Dentro del loop se suma `1` y el `if` calcula el módulo de `current_number` con `2`. Si es `0` (el número es divisible por 2), `continue` le dice a Python que ignore el resto del loop y vuelva al inicio. Si no es divisible por 2, se ejecuta el resto y se imprime el número.

### Avoiding Infinite Loops

Todo `while` necesita una forma de terminar. Por ejemplo, este loop cuenta de 1 a 5:

````python
x = 1
while x <= 5:
    print(x)
    x += 1
````

Pero si por error se omite la línea `x += 1`, el loop **corre para siempre**:

```python
# This loop runs forever!
x = 1
while x <= 5:
    print(x)
```

El valor de `x` parte en `1` y nunca cambia. Entonces la prueba `x <= 5` siempre es `True` y se imprime una serie infinita de `1`.
Todos los programadores escriben un loop infinito de vez en cuando, sobre todo si la condición de salida es sutil. Si un programa se queda pegado, se presiona `Ctrl-C` o se cierra la ventana de salida.

Para evitarlos, hay que **probar cada `while`** y comprobar que se detiene cuando se espera. Si el programa debe terminar al ingresar un valor, se ejecuta y se ingresa ese valor. Si no termina, se revisa cómo se maneja ese valor. Al menos una parte del programa debe poder volver la condición `False` o llegar a un `break`.

> VS Code muestra la salida en una terminal incrustada. Para cancelar un loop infinito hay que hacer clic en el área de salida antes de presionar `Ctrl-C`. En un notebook, se usa el botón *Interrupt*.

## Using a while Loop with Lists and Dictionaries

Hasta ahora se ha trabajado con una sola pieza de información del usuario a la vez. Para llevar la cuenta de muchos usuarios y muchos datos hay que usar listas y diccionarios con los `while`.

Un `for` sirve para recorrer una lista, pero **no se debe modificar una lista dentro de un `for`**, porque Python tiene problemas para llevar la cuenta de los elementos. Para modificar una lista mientras se recorre, se usa un `while`. Así se puede reunir, guardar y organizar mucha información para revisarla después.

### Moving Items from One List to Another

Considerar una lista de usuarios recién registrados en un sitio web, pero aún no verificados. Después de verificarlos, hay que pasarlos a una lista de usuarios confirmados. Una forma es usar un `while` que saque a los usuarios de la lista de no confirmados a medida que se verifican y los agregue a la lista de confirmados:

````python
# Start with users that need to be verified,
#  and an empty list to hold confirmed users.
unconfirmed_users = ['alice', 'brian', 'candace']
confirmed_users = []

# Verify each user until there are no more unconfirmed users.
#  Move each verified user into the list of confirmed users.
while unconfirmed_users:
    current_user = unconfirmed_users.pop()

    print(f"Verifying user: {current_user.title()}")
    confirmed_users.append(current_user)

# Display all confirmed users.
print("\nThe following users have been confirmed:")
for confirmed_user in confirmed_users:
    print(confirmed_user.title())
````

- Se parte con una lista de usuarios no confirmados (Alice, Brian y Candace) y una lista vacía para los confirmados.
- El `while` corre mientras la lista `unconfirmed_users` **no esté vacía** (una lista vacía es `False`).
- Dentro del loop, `pop()` saca al usuario del **final** de `unconfirmed_users`. Como Candace es la última, es la primera en salir: se asigna a `current_user` y se agrega a `confirmed_users`. Luego sigue Brian y después Alice.
- Se simula la confirmación imprimiendo un mensaje de verificación y agregando al usuario a la lista de confirmados. Cuando la lista de no confirmados queda vacía, el loop se detiene y se imprime la de confirmados.

### Removing All Instances of Specific Values from a List

En el Capítulo 3 se usó `remove()` para quitar un valor de una lista. Funcionaba porque el valor aparecía **una sola vez**. Pero si se quieren quitar *todas* las apariciones de un valor, se usa un `while` que corre hasta que ese valor ya no esté en la lista.
Por ejemplo, una lista de mascotas con el valor `'cat'` repetido varias veces:

````python
pets = ['dog', 'cat', 'dog', 'goldfish', 'cat', 'rabbit', 'cat']
print(pets)

while 'cat' in pets:
    pets.remove('cat')

print(pets)
````

Se parte con una lista con varios `'cat'`. Python entra al `while` porque encuentra `'cat'` en la lista al menos una vez, quita la primera aparición, vuelve a la línea del `while` y entra de nuevo mientras siga habiendo `'cat'`. Sale del loop cuando el valor ya no está.

### Filling a Dictionary with User Input

En cada vuelta de un `while` se puede pedir tanta información como se necesite. Este programa de encuesta pide el nombre y una respuesta en cada vuelta, y guarda los datos en un **diccionario**, porque cada respuesta se quiere conectar con un usuario en particular:

````python
responses = {}

# Set a flag to indicate that polling is active.
polling_active = True

while polling_active:
    # Prompt for the person's name and response.
    name = input("\nWhat is your name? ")
    response = input("Which mountain would you like to climb someday? ")

    # Store the response in the dictionary.
    responses[name] = response

    # Find out if anyone else is going to take the poll.
    repeat = input("Would you like to let another person respond? (yes/ no) ")
    if repeat == 'no':
        polling_active = False

# Polling is complete. Show the results.
print("\n--- Poll Results ---")
for name, response in responses.items():
    print(f"{name} would like to climb {response}.")
````

- El programa parte con un diccionario vacío (`responses`) y un flag (`polling_active`) que indica que la encuesta está activa. Mientras `polling_active` sea `True`, el `while` corre.
- Dentro del loop se pide el nombre y una montaña que le gustaría escalar, se guarda en el diccionario (`responses[name] = response`) y se pregunta si seguirá la encuesta. Si responde `yes`, vuelve al `while`; si responde `no`, `polling_active` pasa a `False` y el loop termina.
- Al final, un `for` con `.items()` muestra los resultados.

Con las respuestas `Eric / Denali / yes / Lynn / Devil's Thumb / no`, la salida sería:

```
--- Poll Results ---
Eric would like to climb Denali.
Lynn would like to climb Devil's Thumb.
```

### Comparison: Using a while Loop with Lists and Dictionaries

| | `for` | `while` |
|---|---|---|

| Modificar la lista mientras se recorre | **No** (Python pierde la cuenta) | **Sí** |
| Condición típica | — | `while lista:` (sigue mientras haya elementos) |

Con `pets = ['dog', 'cat', 'dog', 'goldfish', 'cat', 'rabbit', 'cat']`:

| | `pets.remove('cat')` | `while 'cat' in pets: pets.remove('cat')` |
|---|---|---|
| Cuántos `'cat'` quita | solo el primero | todos |
| Resultado | `['dog', 'dog', 'goldfish', 'cat', 'rabbit', 'cat']` | `['dog', 'dog', 'goldfish', 'rabbit']` |
| Cuándo usarlo | el valor aparece una sola vez | el valor puede estar repetido |

| | Mover elementos de una lista a otra | Llenar un diccionario con el usuario |
|---|---|---|
| Estructura de datos | dos listas | un diccionario |

| En cada vuelta | `pop()` + `append()` | `input()` + `responses[name] = response` |
| Termina cuando | la primera lista queda vacía | el usuario responde `no` |

| Si necesito... | Uso |
|---|---|
| Pedir un texto al usuario | `input("prompt ")` |
| Pedir un número | `int(input("prompt "))` |
| Saber si un número es par / múltiplo de otro | `n % 2 == 0` / `n % 10 == 0` |
| Repetir hasta que el usuario quiera salir | `while message != 'quit':` |
| Controlar muchos eventos que terminen el programa | un **flag** (`active = True` ... `active = False`) |
| Salir del loop de inmediato | `break` |
| Saltarse una vuelta del loop | `continue` |
| Un loop que corre hasta un `break` | `while True:` |
| Detener un loop infinito | `Ctrl-C` |
| Mover elementos de una lista a otra | `while lista:` + `pop()` + `append()` |
| Quitar todas las apariciones de un valor | `while valor in lista:` + `remove()` |
| Guardar respuestas de varios usuarios | un diccionario `responses[name] = response` |

| | `while condición` | Flag | `break` | `continue` |
|---|---|---|---|---|
| Qué controla | si el loop sigue | si el programa sigue activo | salir del loop de inmediato | saltar a la siguiente vuelta |
| Se define | en la línea del `while` | una variable al inicio (`active = True`) | dentro del loop, en un `if` | dentro del loop, en un `if` |
| Ejemplo | `while current_number <= 5:` | `while active:` | `if city == 'quit': break` | `if current_number % 2 == 0: continue` |

| Tipo de dato de entrada | Cómo llega | Cómo se usa |
|---|---|---|
| Texto | tal cual (`str`) | directo |
| Número | como `str` | se convierte con `int()` antes de comparar o calcular |
