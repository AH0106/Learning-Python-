## Summary ##

Una **lista** es una colección de items con un orden particular, escrita entre corchetes `[ ]`.
Se accede a sus elementos por **índice**, que parte en **0** (`lista[0]` es el primero) y puede ser negativo (`lista[-1]` es el último).
Se **modifica** un elemento con `lista[i] = valor`, se **agrega** con `.append()` (al final) o `.insert()` (en cualquier posición) y se **elimina** con `del`, `.pop()` o `.remove()`.
Se **ordena** de forma permanente con `.sort()`, de forma temporal con `sorted()`, se invierte con `.reverse()` y se mide con `len()`.
El error más común es `IndexError: list index out of range`.

## Modifyinf, Adding, and Removing Elements ##

### Modifying elements in a List ###

La sintaxis para modificar un elemento en una lista es similar a la que se usa para acceder a un elemento en esta 

Por ejemplo, se tiene una lista de motocicletas y el primer elemento en la lista es 'honda'. 
Se puede cambiar el valor del primer item en la lista despues de que la lista haya sido creada

```python
motorcycle = ['honda', 'yamaha', 'suzuki']
print(motorcycle)

motorcycle[0] = 'ducati'
print(motorcycle)
```

### Adding elements to a List ###

#### Appending Elements to the End of a list ####

La forma mas simple de añadir un elemento a una lista es añadir el item a la lista al final. 


```python
motorcycle = ['Harley Davidson', 'Ducati', 'Kawasaki', 'Yamaha']

print(motorcycle)

motorcycle.append('Suzuki')
print(motorcycle)
```

El medoto .append() hace facil crear listas dinamicas. 

Por ejemplo, se puede empezar con una lista vacia y luego añadirle items usando una serie de .append()

```python
motorcycle = []

motorcycle.append('Harley Davidson')
motorcycle.append('Ducati')
motorcycle.append('Kawasaki')

print(motorcycle)
```

#### Inserting Elements into a list #### 

Se pueden añadir elementos en cualquier posicion de una lista usando el metodo

.insert()

Se hace esto especificando el index del nuevo elemento y el valor del nuevo item 

```python
motorcycle = ['Harley Davidson', 'Ducati', 'Kawasaki', 'Yamaha']

motorcycle.insert(0, 'Suzuki')
print(motorcycle)
```

### Comparison: Modifying and Adding Elements ###

Con `motorcycle = ['honda', 'yamaha', 'suzuki']`:

| | Modificar | `.append()` | `.insert()` |
|---|---|---|---|
| Sintaxis | `lista[0] = 'ducati'` | `lista.append('ducati')` | `lista.insert(0, 'ducati')` |
| Qué hace | **reemplaza** un elemento existente | **agrega** al final | **agrega** en la posición indicada |
| Resultado | `['ducati', 'yamaha', 'suzuki']` | `['honda', 'yamaha', 'suzuki', 'ducati']` | `['ducati', 'honda', 'yamaha', 'suzuki']` |
| ¿Cambia el largo? | No | Sí (+1) | Sí (+1) |
| Cuándo usarlo | cambiar un valor ya guardado | construir listas dinámicas (empezar con `[]` y agregar) | agregar al inicio o al medio |

#### Removing an Item Using the  del Statement ####

Si se sabe la posicion del item que se quiere remover, se puede usar la declaracion $del$

```python
motorcycles = ['Honda', 'Yamaha', 'Suzuki', 'Kawasaki']

print(motorcycles)

del motorcycles[0]
print(motorcycles)
```

#### Removing an Item Using the pop() Method ####

Algunas veces se querra usar el valor de un item después de borrarlo de la lista. 

El metodo .pop() remueve el último item de una lista, pero te deja trabajar con ese item después de removerlo. 

```python
motorcycles = ['Honda', 'Yamaha', 'Suzuki', 'Kawasaki']

print(motorcycles)

popped_motorcycle = motorcycles.pop()

print(motorcycles)
print(popped_motorcycle)
```

Cómo podría ser el metodo .pop() util? 

Imagina que que las motociclestas en la lista son guardadas en orden cronologico, de acuerdo a cuándo fueron adquiridas.

En este caso, se puede usar el motodo .pop() para imprimir una declaración acerca de la ultima motocicleta adquirida. 

```python
motorcycles = ['Honda', 'Yamaha', 'Suzuki', 'Kawasaki']

last_owned = motorcycles.pop()

print(f"The last motorcycle I owned was a {last_owned.title()}.")
```

#### Popping Items from Any Position in a List ####

Se puede usar el metodo pop() para remover un Item en cualquier posición de una lista, incluyendo el index Item que se quiere remover.  

```python
motorcycles = ['Honda', 'Yamaha', 'Suzuki', 'Kawasaki']

first_owned = motorcycles.pop(0)

print(f"The first motorcycle I owned was a {first_owned.title()}.")
```

#### Removing a item by Value ####

Algunas veces no sabes la posicion del valor que quieres eliminar de una lista. Si solo se sabe el valor del item que quieres remover, se puede usar el metodo remove().

Por ejemplo, se quiere remover el valor 'Susuki' de la lista de motocicletas

```python
motorcycles = ['Honda', 'Yamaha', 'Suzuki', 'Kawasaki']

print(motorcycles)


motorcycles.remove('Suzuki')

print(motorcycles)
```

Tambien se puede usar el metodo remove() para trabajar con el valor que a sido removido de una lista. 
Removeremos el valor 'Suzuki' y imprimiremos la razon del porque se removio de la lista

```python
motorcycles = ['Honda', 'Yamaha', 'Suzuki', 'Kawasaki']
print(motorcycles)

too_expensive = 'Suzuki'

motorcycles.remove(too_expensive)

print(motorcycles)

print(f"\nA {too_expensive.title()} is too expensive for me.")
```

### Comparison: Removing Elements ###

Con `motorcycles = ['Honda', 'Yamaha', 'Suzuki', 'Kawasaki']`:

| | `del` | `.pop()` | `.pop(i)` | `.remove(x)` |
|---|---|---|---|---|
| Sintaxis | `del motorcycles[0]` | `motorcycles.pop()` | `motorcycles.pop(0)` | `motorcycles.remove('Suzuki')` |
| ¿Cómo se indica qué eliminar? | por **posición** | siempre el **último** | por **posición** | por **valor** |
| Lista después | `['Yamaha', 'Suzuki', 'Kawasaki']` | `['Honda', 'Yamaha', 'Suzuki']` | `['Yamaha', 'Suzuki', 'Kawasaki']` | `['Honda', 'Yamaha', 'Kawasaki']` |
| ¿Se puede usar el valor eliminado? | No | Sí: devuelve `'Kawasaki'` | Sí: devuelve `'Honda'` | No devuelve nada, pero se puede guardar en una variable antes |
| Cuándo usarlo | se sabe la posición y el valor no importa | se necesita el último elemento (ej: la última moto comprada) | se necesita un elemento de cualquier posición | se conoce el valor pero no la posición |

`.remove()` elimina solo la **primera** aparición del valor. Si el valor está repetido, hay que usar un loop para eliminar todas.

### *Organizing a List* ###

Frecuentemente, vas a querer presentar una lista en un orden particular, otras veces querras preservar el orden original de la lista y otras, querras cambiar el orden original de la lista.

#### Sorting a List Permanently with the sort() Method ####

El metodo *sort()* hacer relativamente facil ordenar una lista. 

Imagina que tienes la lista de unos autos y quieres ordenarla alfabeticamente. 


```python
cars = ['bmw', 'audi', 'toyota', 'subaru']
print(f"\n{cars}")

cars.sort()
print(f"\n{cars}")
```

Tambien se puede ordenar en alfabeticamente pero al revez, pasando el argumento *reverse=True* al metodo *sort()*

```python
cars = ['bmw', 'audi', 'toyota', 'subaru']

print(cars)

cars.sort(reverse=True)
print(f"\n{cars}")
```

### Sorting a List Temporarily with the *sorted()* Function ###

Para mantener el orden original de una lista pero mostrarla/presentarla en orden, 
se puede usar la función *sorted()*.

La función *sorted()* deja mostrar la lista en un orden particular, pero no afecta el orden actual de la lista. 

```python
cars = ['bmw', 'audi', 'toyota', 'honda']

print("Here is the original list")
print(cars)

print("\nHere is the sorted list")
print(sorted(cars))

print("\nHere is the original list again")
print(cars)
```

#### Printing a List in Reverse Order ####

Para revertir el orden original de una lista, se usa el metodo ***reverse()***. 

Si originalmente se guardo en la lista de autos en orden cronologico, de acuerdo a cuando lo adquirimos, 
se puede reordenar la lista de al reves cronologicamente. 

```python
cars = ['bmw', 'audi', 'toyota', 'honda']
print(cars)

cars.reverse()
print(cars)
```

Notar que el metodo ***reverse()*** no ordena alfabeticamente al reves, simplemente reordena la lista de manera inversa.

Este metodo cambia el orden de la lista de manera permanente. Pero se puede volver al orden original volviendo a usar el metodo ***reverse()***

#### *Finding the Length of a List* ####

Se puede saber el tamaño de una lista usando la función ***len()***

```python
cars = ['bmw', 'audi', 'toyota', 'honda']

print(len(cars))
```

### Comparison: Organizing a List ###

Con `cars = ['bmw', 'audi', 'toyota', 'subaru']`:

| | `.sort()` | `.sort(reverse=True)` | `sorted(cars)` | `sorted(cars, reverse=True)` | `.reverse()` | `len(cars)` |
|---|---|---|---|---|---|---|
| Tipo | método | método | función | función | método | función |
| ¿Cambia la lista original? | **Sí**, permanente | **Sí**, permanente | **No** | **No** | **Sí**, permanente | No |
| Resultado | `['audi', 'bmw', 'subaru', 'toyota']` | `['toyota', 'subaru', 'bmw', 'audi']` | `['audi', 'bmw', 'subaru', 'toyota']` | `['toyota', 'subaru', 'bmw', 'audi']` | `['subaru', 'toyota', 'audi', 'bmw']` | `4` |
| Criterio | alfabético | alfabético inverso | alfabético | alfabético inverso | **invierte el orden actual** | cantidad de elementos |

- `.reverse()` **no** ordena alfabéticamente al revés, solo invierte el orden en que está la lista. Para volver al original, se vuelve a usar `.reverse()`.
- No se puede volver al orden original después de `.sort()`; por eso, si se necesita conservar el original, se usa `sorted()`.
