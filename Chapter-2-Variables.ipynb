### Variables and Simple Data Types ###




## Summary ##

Una **variable** es un nombre (una etiqueta) que apunta a un valor, y ese valor se puede cambiar en cualquier momento del programa.
Un **string** es una serie de caracteres entre comillas simples o dobles; se transforma con **métodos** como `.title()`, `.upper()`, `.lower()`, `.strip()` o `.removeprefix()`, y se pueden insertar variables dentro de él con **f-strings**.
Los números son **int** (enteros) o **float** (con punto decimal); con `+ - * / **` se hacen operaciones, y `/` siempre devuelve un float.
También se vio: `\t` y `\n` para espacios en blanco, `_` en números largos, asignación múltiple, constantes en MAYÚSCULAS, comentarios con `#` y el *Zen of Python* (`import this`).
Los errores típicos del capítulo son `NameError` y `SyntaxError`.

## Cheat Sheet ##

### Funciones y métodos ###

| Función / método | Qué hace | Ejemplo | Resultado | ¿Modifica el original? |
|---|---|---|---|---|
| `print(x)` | muestra `x` en pantalla | `print("Hello")` | `Hello` | — |
| `.title()` | cada palabra empieza con mayúscula | `"ada lovelace".title()` | `'Ada Lovelace'` | No |
| `.upper()` | todo en mayúsculas | `"Ada Lovelace".upper()` | `'ADA LOVELACE'` | No |
| `.lower()` | todo en minúsculas | `"Ada Lovelace".lower()` | `'ada lovelace'` | No |
| `.rstrip()` | quita espacios a la **derecha** | `'python '.rstrip()` | `'python'` | No |
| `.lstrip()` | quita espacios a la **izquierda** | `' python'.lstrip()` | `'python'` | No |
| `.strip()` | quita espacios en **ambos lados** | `' python '.strip()` | `'python'` | No |
| `.removeprefix(p)` | quita el prefijo `p` del inicio | `'https://nostarch.com'.removeprefix('https://')` | `'nostarch.com'` | No |
| `.removesuffix(s)` | quita el sufijo `s` del final (ejercicio 2-8) | `'python_notes.txt'.removesuffix('.txt')` | `'python_notes'` | No |
| `f"...{var}..."` | f-string: inserta el valor de una variable en un string | `f"{first_name} {last_name}"` | `'ada lovelace'` | — |
| `import this` | muestra el *Zen of Python* | `import this` | principios de Tim Peters | — |

Ningún método de string cambia el original. Para conservar el resultado hay que reasignarlo: `favorite_language = favorite_language.rstrip()`.

### Reglas para nombrar variables ###

- Solo letras, números y guiones bajos; puede empezar con letra o `_`, **no** con número (`message_1` sí, `1_message` no).
- No se permiten espacios; se usa `_` (`greeting_message`).
- No usar palabras reservadas ni nombres de funciones (por ejemplo `print`).
- Nombres cortos pero descriptivos (`student_name` mejor que `s_n`).
- Cuidado con la `l` minúscula y la `O` mayúscula, que se confunden con `1` y `0`.
- Por ahora, en minúsculas.

## Strings ## 

Un string es una serie de caracteres.

Cualquier cosa dentro de " " es considerada un string en Python. 

Se pueden usar comillas simple ' ' o dobles " " para representar un string. 

El metodo  .title()  cambia cada palabra a formato de titulo, donde cada palabra empieza con con mayuscula.  


```python
name = "andres hormazabal is here"

print(name.title())
```

Se puede cambiar un string a todas las letras en mayuscula o minuscula con

.upper()
.lower()


```python
name = "andres hormazabal"

print(name.upper())
print(name.lower())
```

### Using Variables in Strings ###

En algunos casos, se quiere usar el valor de una variable dentro de un string

Por ejemplo, si se quiere usar 2 variables para representar el primer nombre y el segundo nombre, respectivamente; y luego combinar esos valores para mostrar el nombre completo de alguien:

```python
first_name = "andres"
last_name = "hormazabal"
full_name = f"{first_name} {last_name}"
print(full_name)
```

Para insertar el valor de una variable dentro de un string, hay que colocar la letra f inmediatamente antes de las comillas de apertura.

Hay que colocar llaves alrededor del nombre o cualquier  variables que se quiera usar dentro de un string.

Python va reemplazar cada variable con su valor cuando el strign es mostrado. 

```python
first_name = "andres"
last_name = "hormazabal"
full_name = f"{first_name} {last_name}"
print(f"Hello, {full_name.title()}!")
print(message)
```
