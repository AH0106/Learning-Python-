## Summary

En la **programación orientada a objetos (OOP)** se escriben **clases** que representan cosas y situaciones del mundo real, y se crean **objetos** a partir de esas clases. Crear un objeto a partir de una clase se llama **instanciación**, y se trabaja con **instancias**.
Una clase define el comportamiento general: los **atributos** (información que guarda cada instancia, definida en el método especial `__init__()`) y los **métodos** (acciones que puede hacer). El parámetro `self` conecta cada instancia con sus atributos y métodos.
Los atributos se pueden modificar directamente, con un método o incrementándolos con un método. Una clase puede **heredar** de otra (`class ElectricCar(Car)`) y llamar al `__init__()` de la clase padre con `super()`, sobrescribir métodos del padre, o usar instancias de otras clases como atributos (**composición**).
Las clases se guardan en **módulos** y se importan en otros archivos (`from car import Car`, `import car`, con alias, etc.). También se pueden usar módulos de la **biblioteca estándar**, como `random`, y se siguen convenciones de estilo: `CamelCase` para las clases y docstrings en todas partes.

## Creating and Using a Class

Casi cualquier cosa se puede modelar con clases. Se empieza con una clase simple, `Dog`, que representa a un perro cualquiera (no uno en particular).
¿Qué se sabe de la mayoría de los perros? Todos tienen un nombre y una edad, y la mayoría se sienta (*sit*) y da una vuelta (*roll over*). Esos dos datos (nombre y edad) y esas dos conductas van en la clase, porque son comunes a casi todos los perros. La clase le dice a Python cómo crear un objeto que represente a un perro; luego se usa para crear instancias, cada una representando a un perro específico.

### Creating the Dog Class

Cada instancia creada a partir de `Dog` guarda un nombre y una edad, y puede hacer `sit()` y `roll_over()`:

````python
class Dog:
    """A simple attempt to model a dog."""

    def __init__(self, name, age):
        """Initialize name and age attributes."""
        self.name = name
        self.age = age

    def sit(self):
        """Simulate a dog sitting in response to a command."""
        print(f"{self.name} is now sitting.")

    def roll_over(self):
        """Simulate rolling over in response to a command."""
        print(f"{self.name} rolled over!")
````

- Se define una clase llamada `Dog`. Por convención, los nombres con mayúscula se refieren a clases. No hay paréntesis en la definición porque la clase se crea desde cero. Después va un **docstring** que describe lo que hace la clase.

### The __init__() Method

Una función que es parte de una clase se llama **método**. Todo lo que se aprendió sobre funciones sirve para los métodos; la única diferencia práctica por ahora es la forma de llamarlos.
`__init__()` es un método especial que Python ejecuta automáticamente cada vez que se crea una instancia nueva a partir de la clase. Tiene dos guiones bajos al inicio y dos al final, una convención para que los nombres por defecto de Python no choquen con los propios. Hay que usar **dos a cada lado**: con uno solo el método no se llama automáticamente, y los errores son difíciles de identificar.

Se define `__init__()` con tres parámetros: `self`, `name` y `age`.
- El parámetro **`self`** es obligatorio, y debe ir primero. Es una referencia a la instancia misma, y le da a cada instancia acceso a los atributos y métodos de la clase. Al crear una instancia, Python llama a `__init__()` y le pasa `self` automáticamente, así que solo hay que dar los valores de los otros parámetros.
- Las dos variables del cuerpo de `__init__()` llevan el prefijo `self`. Toda variable con ese prefijo está disponible para todos los métodos de la clase y se puede acceder a ella a través de cualquier instancia. `self.name = name` toma el valor del parámetro `name` y lo asigna a la variable `name` que queda unida a la instancia. Las variables a las que se accede a través de instancias se llaman **atributos**.
- `sit()` y `roll_over()` no necesitan información adicional, así que solo tienen el parámetro `self`. Por ahora solo imprimen un mensaje, pero el concepto se puede extender a casos reales: en un videojuego tendrían el código para animar a un perro; en un robot, moverían un perro robótico.

### Making an Instance from a Class

Una clase es un conjunto de instrucciones sobre cómo crear una instancia. Por ejemplo, para crear un perro específico:

````python
class Dog:
    """A simple attempt to model a dog."""

    def __init__(self, name, age):
        """Initialize name and age attributes."""
        self.name = name
        self.age = age

    def sit(self):
        """Simulate a dog sitting in response to a command."""
        print(f"{self.name} is now sitting.")

    def roll_over(self):
        """Simulate rolling over in response to a command."""
        print(f"{self.name} rolled over!")

my_dog = Dog('Willie', 6)

print(f"My dog's name is {my_dog.name}.")
print(f"My dog is {my_dog.age} years old.")
````

Python llama al `__init__()` de `Dog` con los argumentos `'Willie'` y `6`, que crea una instancia que representa a ese perro con sus atributos `name` y `age`, y la devuelve; esa instancia se asigna a la variable `my_dog`. Una convención útil: un nombre con mayúscula como `Dog` se refiere a una clase, y uno en minúscula como `my_dog` a una instancia.

#### Accessing Attributes

Para acceder a los atributos de una instancia se usa la **notación de punto**: `my_dog.name`. Python busca la instancia `my_dog` y luego el atributo `name` asociado a ella (el mismo que dentro de la clase se llama `self.name`).

#### Calling Methods

Después de crear una instancia, la notación de punto sirve para llamar cualquier método definido en `Dog`:

````python
class Dog:
    """A simple attempt to model a dog."""

    def __init__(self, name, age):
        """Initialize name and age attributes."""
        self.name = name
        self.age = age

    def sit(self):
        """Simulate a dog sitting in response to a command."""
        print(f"{self.name} is now sitting.")

    def roll_over(self):
        """Simulate rolling over in response to a command."""
        print(f"{self.name} rolled over!")

my_dog = Dog('Willie', 6)
my_dog.sit()
my_dog.roll_over()
````

Para llamar a un método se da el nombre de la instancia y el del método, separados por un punto. Cuando se usan nombres descriptivos (`name`, `age`, `sit()`, `roll_over()`) es fácil deducir qué hace un bloque de código, incluso uno que nunca se ha visto.

#### Creating Multiple Instances

Se pueden crear tantas instancias como se necesiten. Se crea un segundo perro, `your_dog`:

````python
class Dog:
    """A simple attempt to model a dog."""

    def __init__(self, name, age):
        """Initialize name and age attributes."""
        self.name = name
        self.age = age

    def sit(self):
        """Simulate a dog sitting in response to a command."""
        print(f"{self.name} is now sitting.")

    def roll_over(self):
        """Simulate rolling over in response to a command."""
        print(f"{self.name} rolled over!")

my_dog = Dog('Willie', 6)
your_dog = Dog('Lucy', 3)

print(f"My dog's name is {my_dog.name}.")
print(f"My dog is {my_dog.age} years old.")
my_dog.sit()

print(f"\nYour dog's name is {your_dog.name}.")
print(f"Your dog is {your_dog.age} years old.")
your_dog.sit()
````

Cada perro es una instancia separada, con su propio conjunto de atributos y capaz de hacer las mismas acciones. Aunque el segundo perro tuviera el mismo nombre y edad, Python igualmente crearía una instancia separada. Se pueden crear tantas instancias como se necesiten, mientras se les dé un nombre de variable único o un lugar único en una lista o diccionario.

## Working with Classes and Instances

Se pueden usar clases para representar muchas situaciones reales. Una vez escrita una clase, se pasa la mayor parte del tiempo trabajando con las instancias creadas a partir de ella. Una de las primeras tareas es modificar los atributos de una instancia en particular: se pueden modificar directamente o escribir métodos que los actualicen de formas específicas.

### The Car Class

Una clase que representa un auto: guarda información sobre el tipo de auto y tiene un método que la resume:

````python
class Car:
    """A simple attempt to represent a car."""

    def __init__(self, make, model, year):
        """Initialize attributes to describe a car."""
        self.make = make
        self.model = model
        self.year = year

    def get_descriptive_name(self):
        """Return a neatly formatted descriptive name."""
        long_name = f"{self.year} {self.make} {self.model}"
        return long_name.title()

my_new_car = Car('audi', 'a4', 2024)
print(my_new_car.get_descriptive_name())
````

`__init__()` recibe `self` primero y luego `make`, `model` y `year`, y los asigna a los atributos de la instancia. `get_descriptive_name()` junta el año, la marca y el modelo en un solo string, para no tener que imprimir cada atributo por separado, y para eso usa `self.make`, `self.model` y `self.year`.

### Setting a Default Value for an Attribute

Al crear una instancia, se pueden definir atributos sin pasarlos como parámetros. Se definen en `__init__()` con un valor por defecto. Se agrega `odometer_reading`, que siempre parte en `0`, y un método `read_odometer()`:

````python
class Car:
    """A simple attempt to represent a car."""

    def __init__(self, make, model, year):
        """Initialize attributes to describe a car."""
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0

    def get_descriptive_name(self):
        """Return a neatly formatted descriptive name."""
        long_name = f"{self.year} {self.make} {self.model}"
        return long_name.title()

    def read_odometer(self):
        """Print a statement showing the car's mileage."""
        print(f"This car has {self.odometer_reading} miles on it.")

my_new_car = Car('audi', 'a4', 2024)
print(my_new_car.get_descriptive_name())
my_new_car.read_odometer()
````

Pocos autos se venden con exactamente 0 millas, así que hace falta una forma de cambiar ese valor.

### Modifying Attribute Values

El valor de un atributo se puede cambiar de **tres** formas: directamente desde la instancia, con un método que lo fija, o con un método que lo incrementa.

#### Modifying an Attribute's Value Directly

La forma más simple es acceder al atributo directamente a través de la instancia. Aquí se fija el odómetro en 23:

````python
class Car:
    """A simple attempt to represent a car."""

    def __init__(self, make, model, year):
        """Initialize attributes to describe a car."""
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0

    def get_descriptive_name(self):
        """Return a neatly formatted descriptive name."""
        long_name = f"{self.year} {self.make} {self.model}"
        return long_name.title()

    def read_odometer(self):
        """Print a statement showing the car's mileage."""
        print(f"This car has {self.odometer_reading} miles on it.")

my_new_car = Car('audi', 'a4', 2024)
print(my_new_car.get_descriptive_name())

my_new_car.odometer_reading = 23
my_new_car.read_odometer()
````

#### Modifying an Attribute's Value Through a Method

A veces conviene tener métodos que actualicen los atributos. En vez de acceder directamente, se le pasa el valor nuevo a un método que hace la actualización internamente. Además se puede agregar lógica para, por ejemplo, **impedir que alguien retroceda el odómetro**:

````python
class Car:
    """A simple attempt to represent a car."""

    def __init__(self, make, model, year):
        """Initialize attributes to describe a car."""
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0

    def get_descriptive_name(self):
        """Return a neatly formatted descriptive name."""
        long_name = f"{self.year} {self.make} {self.model}"
        return long_name.title()

    def read_odometer(self):
        """Print a statement showing the car's mileage."""
        print(f"This car has {self.odometer_reading} miles on it.")

    def update_odometer(self, mileage):
        """
        Set the odometer reading to the given value.
        Reject the change if it attempts to roll the odometer back.
        """
        if mileage >= self.odometer_reading:
            self.odometer_reading = mileage
        else:
            print("You can't roll back an odometer!")

my_new_car = Car('audi', 'a4', 2024)
print(my_new_car.get_descriptive_name())

my_new_car.update_odometer(23)
my_new_car.read_odometer()

my_new_car.update_odometer(10)
````

`update_odometer()` revisa que el valor nuevo tenga sentido antes de modificar el atributo: si el kilometraje dado es mayor o igual al actual, se actualiza; si es menor, avisa que no se puede retroceder el odómetro.

#### Incrementing an Attribute's Value Through a Method

A veces se quiere **sumar** una cantidad en vez de fijar un valor nuevo. Por ejemplo, se compra un auto usado y se recorren 100 millas entre la compra y el registro. Un método `increment_odometer()` recibe esa cantidad y la suma:

````python
class Car:
    """A simple attempt to represent a car."""

    def __init__(self, make, model, year):
        """Initialize attributes to describe a car."""
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0

    def get_descriptive_name(self):
        """Return a neatly formatted descriptive name."""
        long_name = f"{self.year} {self.make} {self.model}"
        return long_name.title()

    def read_odometer(self):
        """Print a statement showing the car's mileage."""
        print(f"This car has {self.odometer_reading} miles on it.")

    def update_odometer(self, mileage):
        """
        Set the odometer reading to the given value.
        Reject the change if it attempts to roll the odometer back.
        """
        if mileage >= self.odometer_reading:
            self.odometer_reading = mileage
        else:
            print("You can't roll back an odometer!")

    def increment_odometer(self, miles):
        """Add the given amount to the odometer reading."""
        self.odometer_reading += miles

my_used_car = Car('subaru', 'outback', 2019)
print(my_used_car.get_descriptive_name())

my_used_car.update_odometer(23_500)
my_used_car.read_odometer()

my_used_car.increment_odometer(100)
my_used_car.read_odometer()
````

Se puede modificar este método para rechazar incrementos negativos, de modo que nadie lo use para retroceder el odómetro.

> Los métodos como estos controlan cómo los usuarios de un programa actualizan valores como el odómetro, pero cualquiera con acceso al programa puede fijar el odómetro en cualquier valor accediendo directamente al atributo. Una seguridad efectiva requiere mucha atención al detalle, además de revisiones básicas como estas.

## Inheritance

No siempre hay que partir desde cero al escribir una clase. Si la clase es una versión especializada de otra que ya se escribió, se puede usar **herencia**: cuando una clase hereda de otra, adopta sus atributos y métodos. La clase original es la **clase padre** (*parent class*) y la nueva es la **clase hija** (*child class*). La hija puede heredar todos o algunos de los atributos y métodos del padre, y además puede definir atributos y métodos propios.

### The __init__() Method for a Child Class

Al escribir una clase basada en otra, normalmente se quiere llamar al `__init__()` de la clase padre, para inicializar los atributos que ese método define y dejarlos disponibles en la hija.
Como ejemplo, un auto eléctrico es un tipo específico de auto, así que se basa en la clase `Car`. Solo hay que escribir el código de lo que es específico de los autos eléctricos. Esta es una primera versión de `ElectricCar`, que hace todo lo que hace `Car`:

````python
class Car:
    """A simple attempt to represent a car."""

    def __init__(self, make, model, year):
        """Initialize attributes to describe a car."""
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0

    def get_descriptive_name(self):
        """Return a neatly formatted descriptive name."""
        long_name = f"{self.year} {self.make} {self.model}"
        return long_name.title()

    def read_odometer(self):
        """Print a statement showing the car's mileage."""
        print(f"This car has {self.odometer_reading} miles on it.")

    def update_odometer(self, mileage):
        """Set the odometer reading to the given value."""
        if mileage >= self.odometer_reading:
            self.odometer_reading = mileage
        else:
            print("You can't roll back an odometer!")

    def increment_odometer(self, miles):
        """Add the given amount to the odometer reading."""
        self.odometer_reading += miles

class ElectricCar(Car):
    """Represent aspects of a car, specific to electric vehicles."""

    def __init__(self, make, model, year):
        """Initialize attributes of the parent class."""
        super().__init__(make, model, year)

my_leaf = ElectricCar('nissan', 'leaf', 2024)
print(my_leaf.get_descriptive_name())
````

- Se parte con `Car`. Cuando se crea una clase hija, la clase padre debe estar en el archivo actual y **antes** de la hija.
- La clase hija, `ElectricCar`, lleva el nombre del padre entre paréntesis en su definición. Su `__init__()` recibe la información necesaria para crear un `Car`.
- La función **`super()`** permite llamar a un método de la clase padre. Esa línea llama al `__init__()` de `Car`, lo que le da a la instancia de `ElectricCar` todos los atributos definidos en ese método. El nombre viene de la convención de llamar a la clase padre *superclass* y a la hija *subclass*.
- Para probar que la herencia funciona, se crea un auto eléctrico con la misma información que se daría para un auto normal. `ElectricCar(...)` llama al `__init__()` de `ElectricCar`, que a su vez llama al de `Car`.

### Defining Attributes and Methods for the Child Class

Una vez que una clase hija hereda de un padre, se pueden agregar los atributos y métodos que la diferencian. Por ejemplo, un atributo específico de los autos eléctricos (la batería) y un método que lo describa:

````python
class Car:
    """A simple attempt to represent a car."""

    def __init__(self, make, model, year):
        """Initialize attributes to describe a car."""
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0

    def get_descriptive_name(self):
        """Return a neatly formatted descriptive name."""
        long_name = f"{self.year} {self.make} {self.model}"
        return long_name.title()

    def read_odometer(self):
        """Print a statement showing the car's mileage."""
        print(f"This car has {self.odometer_reading} miles on it.")

    def update_odometer(self, mileage):
        """Set the odometer reading to the given value."""
        if mileage >= self.odometer_reading:
            self.odometer_reading = mileage
        else:
            print("You can't roll back an odometer!")

    def increment_odometer(self, miles):
        """Add the given amount to the odometer reading."""
        self.odometer_reading += miles

class ElectricCar(Car):
    """Represent aspects of a car, specific to electric vehicles."""

    def __init__(self, make, model, year):
        """
        Initialize attributes of the parent class.
        Then initialize attributes specific to an electric car.
        """
        super().__init__(make, model, year)
        self.battery_size = 40

    def describe_battery(self):
        """Print a statement describing the battery size."""
        print(f"This car has a {self.battery_size}-kWh battery.")

my_leaf = ElectricCar('nissan', 'leaf', 2024)
print(my_leaf.get_descriptive_name())
my_leaf.describe_battery()
````

Se agrega el atributo `self.battery_size` con valor inicial `40`: queda asociado a todas las instancias de `ElectricCar`, pero a ninguna de `Car`. Y un método `describe_battery()`.
No hay límite para especializar `ElectricCar`. Un atributo o método que podría pertenecer a **cualquier** auto, y no solo a uno eléctrico, debe ir en la clase `Car`. Así, quien use `Car` también tendrá esa funcionalidad, y `ElectricCar` solo tendrá el código específico de los vehículos eléctricos.

### Overriding Methods from the Parent Class

Se puede **sobrescribir** (*override*) cualquier método de la clase padre que no calce con lo que se quiere modelar en la hija. Para eso se define en la hija un método con el **mismo nombre** que el del padre; Python ignorará el del padre y usará el de la hija.
Por ejemplo, si `Car` tuviera un método `fill_gas_tank()`, no tendría sentido para un auto eléctrico, así que se sobrescribe:

````python
class Car:
    """A simple attempt to represent a car."""

    def __init__(self, make, model, year):
        """Initialize attributes to describe a car."""
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0

    def get_descriptive_name(self):
        """Return a neatly formatted descriptive name."""
        long_name = f"{self.year} {self.make} {self.model}"
        return long_name.title()

    def read_odometer(self):
        """Print a statement showing the car's mileage."""
        print(f"This car has {self.odometer_reading} miles on it.")

    def update_odometer(self, mileage):
        """Set the odometer reading to the given value."""
        if mileage >= self.odometer_reading:
            self.odometer_reading = mileage
        else:
            print("You can't roll back an odometer!")

    def increment_odometer(self, miles):
        """Add the given amount to the odometer reading."""
        self.odometer_reading += miles

    def fill_gas_tank(self):
        """Fill the gas tank."""
        print("Filling the gas tank...")

class ElectricCar(Car):
    """Represent aspects of a car, specific to electric vehicles."""

    def __init__(self, make, model, year):
        """Initialize attributes of the parent class."""
        super().__init__(make, model, year)

    def fill_gas_tank(self):
        """Electric cars don't have gas tanks."""
        print("This car doesn't have a gas tank!")

my_leaf = ElectricCar('nissan', 'leaf', 2024)
my_leaf.fill_gas_tank()
````

> El ejemplo agrega `fill_gas_tank()` a `Car` solo para poder mostrar la sobrescritura; el libro lo plantea como una suposición ("si `Car` tuviera…").

Cuando se usa herencia, las clases hijas conservan lo que necesitan y sobrescriben lo que no necesitan del padre.

### Instances as Attributes

Al modelar algo del mundo real, se va agregando cada vez más detalle a una clase: una lista creciente de atributos y métodos, y archivos cada vez más largos. En esos casos, se puede notar que parte de una clase se puede escribir como una clase separada: se divide la clase grande en clases más pequeñas que trabajan juntas. Esto se llama **composición**.
Por ejemplo, en `ElectricCar` se estarían agregando muchos atributos y métodos específicos de la batería. Se mueven a una clase aparte, `Battery`, y se usa una instancia de `Battery` como atributo de `ElectricCar`:

````python
class Car:
    """A simple attempt to represent a car."""

    def __init__(self, make, model, year):
        """Initialize attributes to describe a car."""
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0

    def get_descriptive_name(self):
        """Return a neatly formatted descriptive name."""
        long_name = f"{self.year} {self.make} {self.model}"
        return long_name.title()

    def read_odometer(self):
        """Print a statement showing the car's mileage."""
        print(f"This car has {self.odometer_reading} miles on it.")

    def update_odometer(self, mileage):
        """Set the odometer reading to the given value."""
        if mileage >= self.odometer_reading:
            self.odometer_reading = mileage
        else:
            print("You can't roll back an odometer!")

    def increment_odometer(self, miles):
        """Add the given amount to the odometer reading."""
        self.odometer_reading += miles

class Battery:
    """A simple attempt to model a battery for an electric car."""

    def __init__(self, battery_size=40):
        """Initialize the battery's attributes."""
        self.battery_size = battery_size

    def describe_battery(self):
        """Print a statement describing the battery size."""
        print(f"This car has a {self.battery_size}-kWh battery.")

class ElectricCar(Car):
    """Represent aspects of a car, specific to electric vehicles."""

    def __init__(self, make, model, year):
        """
        Initialize attributes of the parent class.
        Then initialize attributes specific to an electric car.
        """
        super().__init__(make, model, year)
        self.battery = Battery()

my_leaf = ElectricCar('nissan', 'leaf', 2024)
print(my_leaf.get_descriptive_name())
my_leaf.battery.describe_battery()
````

- `Battery` no hereda de ninguna otra clase. Su `__init__()` tiene un parámetro opcional, `battery_size`, que vale `40` si no se da otro valor. `describe_battery()` se movió a esta clase.
- En `ElectricCar`, la línea `self.battery = Battery()` crea una instancia nueva de `Battery` (de tamaño 40, porque no se especifica) y la asigna al atributo `self.battery`. Pasa cada vez que se llama a `__init__()`: todo `ElectricCar` tiene ahora una instancia de `Battery`.
- Para describir la batería hay que pasar por el atributo del auto: `my_leaf.battery.describe_battery()`. Python busca la instancia `my_leaf`, su atributo `battery` y llama al método de esa instancia de `Battery`.

Parece mucho trabajo extra, pero ahora se puede describir la batería con todo el detalle que se quiera sin llenar la clase `ElectricCar`. Por ejemplo, un método `get_range()` que informa la autonomía según el tamaño de la batería:

````python
class Car:
    """A simple attempt to represent a car."""

    def __init__(self, make, model, year):
        """Initialize attributes to describe a car."""
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0

    def get_descriptive_name(self):
        """Return a neatly formatted descriptive name."""
        long_name = f"{self.year} {self.make} {self.model}"
        return long_name.title()

    def read_odometer(self):
        """Print a statement showing the car's mileage."""
        print(f"This car has {self.odometer_reading} miles on it.")

    def update_odometer(self, mileage):
        """Set the odometer reading to the given value."""
        if mileage >= self.odometer_reading:
            self.odometer_reading = mileage
        else:
            print("You can't roll back an odometer!")

    def increment_odometer(self, miles):
        """Add the given amount to the odometer reading."""
        self.odometer_reading += miles

class Battery:
    """A simple attempt to model a battery for an electric car."""

    def __init__(self, battery_size=40):
        """Initialize the battery's attributes."""
        self.battery_size = battery_size

    def describe_battery(self):
        """Print a statement describing the battery size."""
        print(f"This car has a {self.battery_size}-kWh battery.")

    def get_range(self):
        """Print a statement about the range this battery provides."""
        if self.battery_size == 40:
            range = 150
        elif self.battery_size == 65:
            range = 225

        print(f"This car can go about {range} miles on a full charge.")

class ElectricCar(Car):
    """Represent aspects of a car, specific to electric vehicles."""

    def __init__(self, make, model, year):
        """
        Initialize attributes of the parent class.
        Then initialize attributes specific to an electric car.
        """
        super().__init__(make, model, year)
        self.battery = Battery()

my_leaf = ElectricCar('nissan', 'leaf', 2024)
print(my_leaf.get_descriptive_name())
my_leaf.battery.describe_battery()
my_leaf.battery.get_range()
````

`get_range()` hace un análisis simple: si la capacidad es de 40 kWh, la autonomía es de 150 millas; si es de 65 kWh, de 225. De nuevo, hay que llamarlo a través del atributo `battery` del auto.

### Importing a Single Class

Se crea un módulo con solo la clase `Car`. Aquí aparece un problema de nombres: ya existía un archivo `car.py` en el capítulo, pero este módulo debe llamarse `car.py` porque contiene código que representa un auto. Se resuelve guardando la clase `Car` en un módulo `car.py` (reemplazando al anterior); desde ahora, el programa que lo use necesita un nombre de archivo más específico, como `my_car.py`.

````python
# car.py
"""A class that can be used to represent a car."""

class Car:
    """A simple attempt to represent a car."""

    def __init__(self, make, model, year):
        """Initialize attributes to describe a car."""
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0

    def get_descriptive_name(self):
        """Return a neatly formatted descriptive name."""
        long_name = f"{self.year} {self.make} {self.model}"
        return long_name.title()

    def read_odometer(self):
        """Print a statement showing the car's mileage."""
        print(f"This car has {self.odometer_reading} miles on it.")

    def update_odometer(self, mileage):
        """
        Set the odometer reading to the given value.
        Reject the change if it attempts to roll the odometer back.
        """
        if mileage >= self.odometer_reading:
            self.odometer_reading = mileage
        else:
            print("You can't roll back an odometer!")

    def increment_odometer(self, miles):
        """Add the given amount to the odometer reading."""
        self.odometer_reading += miles
````

Se incluye un **docstring a nivel de módulo** que describe brevemente su contenido: hay que escribir uno para cada módulo. Luego, en otro archivo, `my_car.py`, se importa la clase `Car` y se crea una instancia:

````python
# my_car.py
from car import Car

my_new_car = Car('audi', 'a4', 2024)
print(my_new_car.get_descriptive_name())

my_new_car.odometer_reading = 23
my_new_car.read_odometer()
````

```

This car has 23 miles on it.
```

El `import` le dice a Python que abra el módulo `car` e importe la clase `Car`. Ahora se puede usar como si estuviera definida en el archivo. Importar clases es una forma efectiva de programar: el archivo principal queda limpio y fácil de leer, la lógica queda en archivos separados, y cuando las clases ya funcionan se pueden dejar tranquilas para concentrarse en la lógica de más alto nivel del programa principal.

| Una clase | `from car import Car` | `Car('ford', 'mustang', 2024)` | la forma más simple |

| Módulo completo | `import car` | `car.Car(...)`, `car.ElectricCar(...)` | código fácil de leer; no hay conflictos de nombres |
| Todas las clases | `from car import *` | `Car(...)` | **no recomendado**: no se sabe qué clases se usan y puede haber conflictos |
| Alias de una clase | `from electric_car import ElectricCar as EC` | `EC('nissan', 'leaf', 2024)` | acorta nombres largos |
| Alias de un módulo | `import electric_car as ec` | `ec.ElectricCar(...)` | |
| Módulo dentro de un módulo | `from car import Car` dentro de `electric_car.py` | — | cuando una clase depende de otra de otro módulo |

| | Todo en un archivo | Clases en módulos |
|---|---|---|
| Largo del archivo principal | crece mucho | corto y limpio |
| Dónde está la lógica | mezclada | en archivos separados |
| Cuándo usarlo | al empezar, mientras se prueba | cuando las clases ya funcionan |

## The Python Standard Library

La **biblioteca estándar de Python** es un conjunto de módulos que viene con toda instalación de Python. Con un conocimiento básico de funciones y clases ya se pueden usar módulos escritos por otros programadores: se puede usar cualquier función o clase de la biblioteca estándar con un simple `import` al inicio del archivo.
Un módulo útil para modelar muchas situaciones reales es `random`. Una función interesante es `randint()`, que recibe dos enteros y devuelve un entero aleatorio entre ellos (incluyéndolos):

````python
from random import randint

print(randint(1, 6))
````

Otra función útil es `choice()`, que recibe una lista o tupla y devuelve un elemento elegido al azar:

````python
from random import choice

players = ['charles', 'martina', 'michael', 'florence', 'eli']
first_up = choice(players)
print(first_up)
````

(Los resultados cambian cada vez que se ejecuta la celda.)

> El módulo `random` no debe usarse para aplicaciones relacionadas con seguridad, pero sirve para muchos proyectos divertidos e interesantes. También se pueden descargar módulos de fuentes externas; se verán varios ejemplos en la Parte II.

### Comparison: The Python Standard Library

| | `randint(a, b)` | `choice(secuencia)` |
|---|---|---|
| Recibe | dos enteros | una lista o tupla |
| Devuelve | un entero entre `a` y `b` (ambos incluidos) | un elemento de la secuencia, al azar |
| Ejemplo | `randint(1, 6)` → `3` (un dado) | `choice(players)` → `'florence'` |
| Se importa con | `from random import randint` | `from random import choice` |

## Styling Classes

Algunos puntos de estilo relacionados con las clases, a medida que los programas se complican:

- **Nombres de clases:** en `CamelCase`. Se escribe con mayúscula la primera letra de cada palabra y sin guiones bajos. Los nombres de **instancias y módulos** van en minúsculas, con guiones bajos entre palabras.
- **Docstrings:** toda clase debe tener un docstring inmediatamente después de su definición, con una breve descripción de lo que hace, siguiendo el mismo formato que en las funciones. Cada módulo también debe tener un docstring que describa para qué sirven sus clases.
- **Líneas en blanco:** sirven para organizar el código, pero sin exceso. Dentro de una clase se usa una línea en blanco entre métodos, y dentro de un módulo, dos líneas en blanco para separar clases.
- **Orden de los `import`:** si hay que importar un módulo de la biblioteca estándar y uno propio, primero va el de la biblioteca estándar; luego una línea en blanco y el `import` del módulo propio. Así es más fácil ver de dónde viene cada módulo.

### Comparison: Styling Classes

| | Se escribe en | Ejemplo |
|---|---|---|
| Clases | `CamelCase` | `ElectricCar`, `Dog` |
| Instancias | minúsculas con guiones bajos | `my_leaf`, `your_dog` |
| Módulos | minúsculas con guiones bajos | `electric_car.py`, `car.py` |
| Métodos y atributos | minúsculas con guiones bajos | `get_descriptive_name()`, `odometer_reading` |

| Separador | Líneas en blanco |
|---|---|
| Entre métodos de una clase | 1 |
| Entre clases de un módulo | 2 |

| Si necesito... | Uso |
|---|---|
| Representar una cosa del mundo real | una clase (`class Nombre:`) |
| Dar valores iniciales a cada instancia | `__init__()` con parámetros y `self.atributo = ...` |
| Un atributo que siempre parte con el mismo valor | asignarlo en `__init__()` sin parámetro (`self.odometer_reading = 0`) |
| Cambiar un atributo con reglas | un método (`update_odometer()`) |
| Sumar a un atributo | un método con `+=` (`increment_odometer()`) |
| Una clase que es una versión especializada de otra | herencia: `class Hija(Padre):` + `super().__init__()` |
| Cambiar un comportamiento heredado | sobrescribir el método con el mismo nombre |
| Una parte de una clase que crece mucho | composición: otra clase usada como atributo |
| Mantener limpio el archivo principal | guardar las clases en módulos e importarlas |
| Importar varias clases | `from modulo import A, B` o `import modulo` |
| Acortar un nombre largo al importar | alias con `as` |
| Números o elecciones al azar | `random`: `randint()` y `choice()` |
