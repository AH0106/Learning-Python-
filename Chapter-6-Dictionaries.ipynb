## Summary

Un **diccionario** (`{ }`) es una colección de pares **key-value**: cada llave está conectada a un valor, y se usa la llave para obtenerlo (`alien_0['color']`).
Los diccionarios son dinámicos: se agregan pares con `d['key'] = value`, se modifican reasignando la llave y se eliminan con `del d['key']`.
Para evitar un `KeyError` al pedir una llave que quizás no existe, se usa `.get(key, default)`.
Se recorren con `.items()` (llaves y valores), `.keys()` (solo llaves) o `.values()` (solo valores); con `sorted()` se recorren en orden, y con `set()` se eliminan repetidos.
El **nesting** permite guardar varios diccionarios en una lista, una lista dentro de un diccionario o un diccionario dentro de otro.

## Cheat Sheet

### Funciones, métodos y sintaxis

| Quiero... | Sintaxis | Ejemplo | Resultado | Notas |
|---|---|---|---|---|
| Crear un diccionario | `{key: value, ...}` | `alien_0 = {'color': 'green', 'points': 5}` | | pares separados por comas; llave y valor por `:` |
| Crear uno vacío | `{}` | `alien_0 = {}` | `{}` | se llena después con `d[key] = value` |
| Acceder a un valor | `d[key]` | `alien_0['color']` | `'green'` | `KeyError` si la llave no existe |
| Acceder con valor por defecto | `d.get(key, default)` | `alien_0.get('points', 'No point value assigned.')` | el valor, o el *default* si no existe | sin segundo argumento devuelve `None` |
| Agregar un par | `d[new_key] = value` | `alien_0['x_position'] = 0` | agrega el par | se mantiene el orden en que se agregan |
| Modificar un valor | `d[key] = new_value` | `alien_0['color'] = 'yellow'` | cambia el valor | |
| Eliminar un par | `del d[key]` | `del alien_0['points']` | elimina llave y valor | es permanente |
| Recorrer llaves y valores | `d.items()` | `for name, language in favorite_languages.items():` | cada par | se pueden usar los nombres que se quiera (`k, v`) |
| Recorrer las llaves | `d.keys()` | `for name in favorite_languages.keys():` | cada llave | recorrer el diccionario directamente hace lo mismo |
| Recorrer las llaves en orden | `sorted(d.keys())` | `for name in sorted(favorite_languages.keys()):` | llaves ordenadas | por defecto se recorre en el orden de inserción |
| Recorrer los valores | `d.values()` | `for language in favorite_languages.values():` | cada valor | puede haber repetidos |
| Valores sin repetir | `set(d.values())` | `set(favorite_languages.values())` | valores únicos | un *set* no guarda un orden |
| Crear un set | `{a, b, c}` | `{'python', 'rust', 'python', 'c'}` | `{'rust', 'python', 'c'}` | llaves **sin** pares key-value |
| Revisar si existe una llave | `key in d.keys()` / `key not in d.keys()` | `'erin' not in favorite_languages.keys()` | `True` / `False` | `.keys()` devuelve una secuencia con todas las llaves |
| Cantidad de elementos de un valor-lista | `len(...)` | `len(languages)` | un número | útil para cambiar el mensaje si hay más de uno |

### Nesting

| Estructura | Cómo se escribe | Ejemplo | Cuándo usarla |
|---|---|---|---|
| Lista de diccionarios | `[dict_1, dict_2, ...]` | `aliens = [alien_0, alien_1, alien_2]` | muchos objetos con la misma estructura (aliens, usuarios) |
| Lista dentro de un diccionario | `{'key': [a, b]}` | `pizza = {'crust': 'thick', 'toppings': ['mushrooms', 'extra cheese']}` | más de un valor asociado a una llave |
| Diccionario dentro de un diccionario | `{'key': {...}}` | `users = {'aeinstein': {'first': 'albert', 'last': 'einstein', 'location': 'princeton'}}` | información detallada por cada llave |

- Para entrar en una estructura anidada se encadenan las llaves/índices: `pizza['toppings']`, `user_info['first']`.
- Una lista de diccionarios se puede generar con un loop (`range(30)` + `append`) y recortar con slices (`aliens[:5]`).
- Es mejor que todos los diccionarios de una lista tengan la **misma estructura**.
- No se debería anidar a demasiada profundidad: suele haber una forma más simple.

# Working with Dictionaries

Un `diccionario` en Python es una colección de pares `key-value`. 

Cada **llave** esta conectada a un **valor**,  y se puede usar una **llave** para acceder al **valor** asociado a esa llave. 

En Python, un diccionario esta entre llaves `({})`, con una serie de pares `key-value` dentro de las llaves. 


## Accesing Values in a Dictionary

Para obtener el valor asociado a una llave, hay que dar el nombre del diccionario y hay que colocar la llave dentro de unos `[ ]`


```python
alien_0 = {
    'color': 'green'
}

print(alien_0['color'])
```

Se puede acceder tanto al color del alien como al valor del puntaje de `alien_0`. Si un jugador dispara a este alien, se puede mirar cuantos puntos debería ganar usando un codigo como este:

```python
alien_0 = {
    'color': 'green',
    'points': 5
}

new_points = alien_0['points']
print(f"You just earned {new_points} points")
```

## Adding New Key-Value Pairs

Los diccionarios son estructuras dinámicas, y se pueden añadir nuevos pares `key-value` a un diccionario en cualquier momento. 

Para añadir un nuevo par key-value, hay que dar el nombre del diccionario seguido por una nueva key en `[ ]`, seguido del nuevo valor.

```python
alien_0 = {
    'color': 'green',
    'points': 5
}

alien_0['x_position'] = 0
alien_0['y_position'] = 25

print(alien_0)
```

## Starting with an Empty Dictionary

Algunas veces es conveniente o necesario empezar con un diccionario vacio, y luego añadir cada nuevo item. 

Para empezar a llenar un diccionario vacio, hay que definir un diccionario con unos corchetes vacios `{ }`, y luego añadir cada par `key-value` en su propia linea. 

```python
alien_0 = {}

alien_0['color'] = 'red'
alien_0['points'] = 5

print(alien_0)
```

## Modifying Values in a Dictionary

Para modificar el valor en un diccionario, hay que dar el nombre del diccionario con la `key` en `[ ]` y luego el valor que se quiere asociar a esa key. 

```python
alien_0 = {
    'color': 'red',
    'points': 5
}

print(f"The alien color is {alien_0['color']}")

alien_0['color'] = 'yellow'
print(f"The alien now is {alien_0['color']}")
```

```python
alien_0 = {
    'x_position': 0,
    'y_position': 25, 
    'speed': 'medium'
}
print(f"Original position: {alien_0['x_position']}, {alien_0['y_position']}")

# Move the alien to the right.
# Determine how far to move the alien based on its current speed. 

if alien_0['speed'] == 'slow':
    x_increment = 1
elif alien_0['speed'] == 'medium':
    x_increment = 2
else:
    #This must be a fast alien
    x_increment = 3

# The new position is the old position plus the increment.
alien_0['x_position'] = alien_0['x_position'] + x_increment

print(f"New position: {alien_0['x_position']}")
```

## Removing Key-Value Pairs

Cuando no se necesita una pieza de información que esta guardada en un diccionario, se puede usar la declaración `del` para remover completamente un par `key-value`.

Todo `del` necesita el nombre del diccionario y la llave que se quiere remover. 

```python
alien_0 = { 
    'color': 'red',
    'points': 5
}
print(alien_0)

del alien_0['points']
print(alien_0)
```

## Using gey() to access Values

Se puede usar el metodo `get()` para configurar un valor por default, que sera retornado si la llave pedida no existe.

El metodo `get()` requiere una llave como primer argumento. Como segundo argumento opcional, se puede pasar el valor que retorna en caso de que la llave no exista. 



```python
alien_0 = {
    'color': 'red',
    'speed': 'slow'
}

point_value = alien_0.get('points', 'No point value assigned')

print(point_value)
```

## Looping Through a Dictionary

### Looping Through All Key-Values Pairs

Se considerar un nuevo diccionario diseñado para guardar información acerca de un usuario en un sitio web

```python
user= {
    'username': 'efermi',
    'first': 'enrico',
    'last': 'fermi'
}

for key, value in user.items():
    print(f"\nKey: {key}")
    print(f"Value: {value}")
```

Para escribir un `for loop` para un diccionario:
> hay que crear nombres para 2 variables que mantendran la key y el valor de cada par `key-value`.
> La segunda mitad del for loop incluye el nombre del diccionario seguido por el metodo `items()`, que retorna una secuencia de pares `key-value`. 


### Looping Through All the Keys in a Dictionary

El metodo `keys()` es útil cuando no se necesita trabajar con todos los valores en un diccionario.  

```python
favorite_lenguages= {
    'jen': 'python',
    'sarah': 'c',
    'edward': 'rust',
    'phil': 'python'
}

for name in favorite_lenguages.keys():
    print(name.title())
```

Se puede acceder al valor asociado de cualquier llave dentro de un loop, usando la llave actual. 


```python
favorite_lenguages= {
    'jen': 'python',
    'sarah': 'c',
    'edward': 'rust',
    'phil': 'python'
}


friends = ['phil', 'sarah']

for name in favorite_lenguages.keys():
    print(f"Hi {name.title()}")

    if name in friends:
        language = favorite_lenguages[name].title()
        print(f"\t{name.title()}, I see you love {language}")
```

Primero se hace una lista de los amigos a los cuales se les quiere imprimir un mensaje.

Dentro del loop, se imprime el nombre de cada persona.

Luego, se revisa si el `name` con el que se esta trabajando esta en la lista de `friends`. Si es asi, se determina el lenguaje favorito de la persona usando el nombre del diccionario y el valor correspondiente de la llave `name`.  

### Looping Through a Dictionary's Keys in a particular order.

Algunas veces, se querra recorrer un diccionario en un orden específico. 

Una manera de hacer esto es ordenar las llaves a medida que van siendo retornadas por el for loop. 

Se puede usar la funcion `sorted()` para obtener una copia de las llaves en orden. 

```python
favorite_lenguages= {
    'jen': 'python',
    'sarah': 'c',
    'edward': 'rust',
    'phil': 'python'
}

for name in sorted(favorite_lenguages.keys()):
    print(f"{name.title()}, thank you for taking the pool")
```

### Looping Through All The Values in a Dictionary

Si es de mayor interes los valores que estan guardados en un diccionario, se puede usar el metodo `values()`para retornar la secuencia de valores sin ninguna llave.  

```python
favorite_lenguages= {
    'jen': 'python',
    'sarah': 'c',
    'edward': 'rust',
    'phil': 'python'
}

print("the following lenguages have been mentioned: ")

for lenguage in favorite_lenguages.values():
    print(lenguage.title())
```

Para ver los valores pero sin repetición, se puede usar un `set`.

Un `set` es una colección en la que cada item debe ser único.

```python
favorite_lenguages= {
    'jen': 'python',
    'sarah': 'c',
    'edward': 'rust',
    'phil': 'python'
}

print("the following lenguages have been mentioned: ")
for lenguage in set(favorite_lenguages.values()):
    print(lenguage.title())
```

## Nesting 

Algunas veces se querrá guardar multiples diccionarios en una lista, o una lista de items como valor en un diccionario.

Esto es llamado `nesting`(anidación).

Se puede anidar **un diccionario dentro de una lista**, **una lista de items dentro de un diccionario**, o un diccionario dentro de otro diccionario.


```python
alien_0 = {
    'color':'green',
    'points': 10
}

alien_1 = {
    'color':'yellow',
    'points': 5
}

alien_2 = {
    'color':'black',
    'points': 20
}

aliens = [alien_0, alien_1, alien_2]

for alien in aliens: 
    print(alien)
```

Un ejemplo más realista involucraría más de tres aliens, haciendo un código que automaticamente genere cada alien.

```python
# Make an empty list for storing aliens.
aliens = []

# Make 30 green aliens
for alien_number in range(30):
    new_alien = {
        'color':'green',
        'points': 10,
        'speed': 'slow'
    }
    aliens.append(new_alien)


# Show the first 5 aliens.
for alien in aliens[:5]:
    print(alien)
print("---")

# Show how many aliens have been created.
print(f"Total number of aliens: {len(aliens)}")
```

Estos aliens tienen todos las mismas características.
Si se quiere cambiar el color, se puede usar un `for` loop y una declaración `if` para cambiar el color de los aliens.

```python
# Make an empty list for storing aliens.
aliens = []

# Make 30 green aliens
for alien_number in range(30):
    new_alien = {
        'color':'green',
        'points': 5,
        'speed': 'slow'
    }
    aliens.append(new_alien)

# Cambiar caracteristicas de los primeros 3 aliens
for alien in aliens[:3]:
    if alien['color'] == 'green':
        alien['color'] = 'yellow'
        alien['speed'] = 'medium'
        alien['points'] = 12


# Show the first 10 aliens.
for alien in aliens[:10]:
    print(alien)
print("---")

# Show how many aliens have been created.
print(f"Total number of aliens: {len(aliens)}")
```

## A list in a Dictionary

Algunas veces es útil colocar una lista dentro de un diccionario.

Imagina cómo podrías describir una pizza que alguien esta ordenando. 

Con solo una lista, todo lo que se podría guardar sería una lista de los ingredientes.

Con un diccionario, una lista de ingredientes puede ser solo un aspecto de la pizza que se esta describiendo.

```python
# Store information about a pizza being ordered.
pizza = {
    'crust': 'thick',
    'toppings': ['mushrooms', 'extra cheese'],
}

# Summarize the order 
print(f"You ordered a {pizza['crust']} - crust pizza" " with the following toppings: ")
for topping in pizza['toppings']:
    print(f"\t{topping}")
```

```python
favorite_lenguages= {
    'jen': ['python', 'rust'],
    'sarah': ['c'],
    'edward': ['rust', 'g'],
    'phil': ['python','haskel']
}

for name, languages in favorite_lenguages.items():
    print(f"{name.title()}'s favorite lenguage are:")
    for language in languages:
        print(f"\t{language.title()}")
```

## A Dictionary in a Dictionary

Se puede anidar un diccionario dentro de un diccionario, pero el codigo se puede volver complicado. 

Si se tienen muchos usuarios en un App, cada uno con un nombre de usuario unico, se pueden usar los nombre de usuarios como llaves. 

Luego, puedes guardar la información de cada usuario usando un diccionario como el valor asociado con su nombre de usuario. 

```python
users = {
    'aeinstein': {
        'first': 'albert',
        'last': 'einstein',
        'location': 'princeton',
        },
    'mcurie': {
        'first': 'marie',
        'last': 'curie',
        'location': 'paris',
        },
    }

for username, user_info in users.items():
    print(f"\nUsername: {username}")
    full_name = f"{user_info['first']} {user_info['last']}"
    location = user_info['location']

    print(f"\tFull name: {full_name.title()}")
    print(f"\tLocation: {location.title()}")
```
