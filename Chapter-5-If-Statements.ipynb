## Summary

Un `if` ejecuta un bloque de código solo si una **conditional test** (una expresión que da `True` o `False`) se cumple; el bloque va **indentado**, igual que en un `for`.
Las pruebas comparan valores con `==`, `!=`, `<`, `<=`, `>`, `>=`, se combinan con `and` / `or` y revisan listas con `in` / `not in`.
Según cuántas condiciones haya, se usa un `if` simple, un `if-else` (siempre se ejecuta una de dos acciones) o una cadena `if-elif-else` (se ejecuta **solo un** bloque).
Cuando hay que revisar **todas** las condiciones, se usa una serie de `if` independientes.
Combinados con listas, los `if` permiten tratar de forma distinta ciertos valores dentro de un `for` y comprobar si una lista está vacía antes de recorrerla.

## Cheat Sheet

### Conditional tests

| Quiero comprobar... | Operador | Ejemplo | Resultado |
|---|---|---|---|
| Igualdad | `==` | `car = 'bmw'` → `car == 'bmw'` | `True` |
| Desigualdad | `!=` | `requested_topping != 'anchovies'` | `True` si son distintos |
| Menor que | `<` | `19 < 21` | `True` |
| Menor o igual que | `<=` | `19 <= 21` | `True` |
| Mayor que | `>` | `19 > 21` | `False` |
| Mayor o igual que | `>=` | `19 >= 21` | `False` |
| Igualdad ignorando mayúsculas | `.lower()` + `==` | `'Audi'.lower() == 'audi'` | `True` (el valor original no cambia) |
| Que se cumplan **ambas** | `and` | `age_0 >= 21 and age_1 >= 21` | `True` solo si las dos son `True` |
| Que se cumpla **al menos una** | `or` | `age_0 >= 21 or age_1 >= 21` | `False` solo si las dos son `False` |
| Un valor está en una lista | `in` | `'mushrooms' in requested_toppings` | `True` / `False` |
| Un valor **no** está en una lista | `not in` | `user not in banned_users` | `True` / `False` |
| Una lista no está vacía | nombre de la lista | `if requested_toppings:` | `True` si tiene al menos un elemento; una lista vacía es `False` |

- Un `=` **asigna** (es una orden: *"guarda este valor"*); un `==` **pregunta** (*"¿son iguales?"*).
- La igualdad distingue mayúsculas de minúsculas: `'Audi' == 'audi'` es `False`.
- Un valor `True` / `False` es un **Boolean value**; una conditional test también se llama **Boolean expression**. Sirven para guardar estados: `game_active = True`, `can_edit = False`.
- Los paréntesis son opcionales y mejoran la lectura: `(age_0 >= 21) and (age_1 >= 21)`.

### Estructuras if

| Estructura | Sintaxis | Cuándo usarla | Qué ocurre |
|---|---|---|---|
| `if` simple | `if prueba:` + bloque indentado | una condición, una acción | si es `False`, no pasa nada |
| `if-else` | `if prueba:` ... `else:` | dos casos posibles | **siempre** se ejecuta uno de los dos bloques |
| `if-elif-else` | `if` ... `elif` ... `else` | más de dos casos | se ejecuta **solo el primer** bloque cuya prueba sea `True`; Python salta el resto |
| Varios `elif` | `if` ... `elif` ... `elif` ... `else` | muchos rangos o casos | igual que el anterior; se pueden usar tantos `elif` como se necesiten |
| Sin `else` | `if` ... `elif` ... `elif` | hay una condición final específica | cada bloque necesita pasar su propia prueba; evita que el `else` capture datos inválidos |
| `if` independientes | `if` ... `if` ... `if` | pueden cumplirse **varias** condiciones | se evalúan **todas** las pruebas |

Regla: si solo debe ejecutarse **un** bloque, `if-elif-else`; si pueden ejecutarse **varios**, una serie de `if` independientes.

### if con listas

| Quiero... | Patrón | Ejemplo |
|---|---|---|
| Tratar distinto un valor especial | `for` + `if` / `else` | `for car in cars:` → `if car == 'bmw': print(car.upper())` `else: print(car.title())` |
| Avisar de un elemento no disponible | `for` + `if` / `else` | `if requested_topping == 'green peppers': ...` |
| Comprobar que la lista no esté vacía | `if lista:` antes del `for` | `if requested_toppings:` ... `else: print("Are you sure you want a plain pizza?")` |
| Revisar una lista contra otra | `for` + `in` con una segunda lista | `if requested_topping in available_toppings:` |

### Estilo

| Regla | Ejemplo |
|---|---|
| Un espacio a cada lado de los operadores de comparación | `if age < 4:` mejor que `if age<4:` |
